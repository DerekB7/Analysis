$T \text{ is compact }\iff T^*T\text{ is compact }$  

$T \text{ is compact }\implies T^*T\text{ is compact }$  
The forward direction follows immediately from ex 4.6

$T \text{ is compact }\impliedby T^*T\text{ is compact }$  
My usual argument is to apply spectral theorem for compact operators to $T^*T$ [Theorem 4.24] and square root each positive eigenvalue to get $\sqrt{T^*T}$ which can be approximated by [bounded] finite rank operators hence it is compact [Theorem 4.10] and by Polar decomposition [ref Brezis] we may write $T=U\sqrt{T^*T}$ where $U$ is partial isometry and $T$ is compact by ex 4.6.    

*However there is a simpler, more direct argument.*   
If $T\in B(H_1,H_2)$ [where $H_1,H_2$ are Hilbert] then let $x_k\in H_1$  be an arbitrary bounded sequence, i.e. $\Vert x_k\Vert \leq C\lt \infty$ and since $T^*T$ is compact then there is a convergent subsequence $\big\{T^*Tx_{k_j}\big\}$ which means it is Cauchy, so by Cauchy-Schwarz and triangle inequality     
$\big\Vert T\big(x_{k_j}-x_{k_m}\big)\big\Vert^2= \big\langle T^*T\big(x_{k_j}-x_{k_m}\big),\big(x_{k_j}-x_{k_m}\big)\big\rangle \leq \big\Vert T^*T\big(x_{k_j}-x_{k_m}\big)\big\Vert \cdot \big\Vert x_{k_j}-x_{k_m}\big\Vert\leq  \big\Vert T^*Tx_{k_j}-T^*Tx_{k_m}\big\Vert\cdot 2C\to 0$  
$\implies \big\{Tx_{k_j}\big\}$ is Cauchy hence converges by completeness of $H_2$.  
$\implies T$ is compact since an arbitrary bounded sequence has a convergent subsequence under the image of $T$  


*Commentary on Hilbert-Schmidt criterion:*   

Theorem 4.15 states the result for separable Hilbert spaces.  However the criterion formally holds for all Hilbert spaces.  Using e.g. ex 3.1, suppose we have some orthonormal basis $\big\{e_\alpha\big\}$ with $\alpha \in \mathcal I$ and we know the sum of squares under the image of $T$ is finite, i.e.  

$$ \sum_{\alpha \in \mathcal I} \big\Vert T e_\alpha\big\Vert^2:=\int \big\Vert T e_\alpha\big\Vert^2 d\mu \lt\infty$$
where the integral is over $\mathcal I$ with counting measure $\mu$.  Assuming one knows of the square-root, [ref either chapter 5 materials of MacCluer or 'square_root_power_series_approach.ipynb'] then the above argument yields  $T \text{ is compact }\iff T^*T\text{ is compact }\iff \sqrt{T^*T} \text{ is compact }$  so we can assume **WLOG** the nicer case of a positive self-adjoint operator $T$. And the fact that the above 'sum' is finite implies $\mathcal I_k=\big\{\alpha:\big\Vert T e_\alpha\big\Vert\geq \frac{1}{k}\big\}$ is a finite set for $k \in \mathbb N$, i.e. consider characteristic function $\mathcal X_{\mathcal I_k}$ and  

$$\frac{1}{k^2}\int \mathcal X_{\mathcal I_k}d\mu\leq \int \mathcal X_{\mathcal I_k}\big\Vert T e_\alpha\big\Vert^2 d\mu \leq \int \big\Vert T e_\alpha\big\Vert^2 d\mu\lt\infty\implies \mu\big(\mathcal X_{\mathcal I_k}\big)= \int \mathcal X_{\mathcal I_k}d\mu\lt \infty$$  

This then in turn implies $\big(\ker T\big)^\perp$ is a separable Hilbert space, which is $T$-invariant, so we may restrict our analysis to $T\vert_{(\ker T)^\perp}$ i.e. a bounded operator on a separable Hilbert space exhibiting the Hilbert-Schmidt property hence it is compact per Theorem 4.15 and $T$ is as well.  See the initial analysis for the 'corollary' to ex 4.19 for more information.  



*Another note on Hilbert-Schmidt operators:*  

To avoid linguistic and other issues here I stick to separable Hilbert spaces.  What ex 4.19 and its converse tells us is that $T\in B(\mathscr H)$ is compact *iff* $Te_k\to 0$ for arbitrary orthonormal sequence $\big\{e_k\big\}_{k=1}^\infty$.  Now letting $\big\{e_k\big\}_{k=1}^\infty$ be an orthonormal basis then if $T$ is Hilbert-Schmidt, i.e. $\big\{Te_k\big\}_{k=1}^\infty$ is square-summable then certainly $Te_k\to 0$ so this seems like a much more onerous requirement.  However for Hilbert-Schdmit operators we need only check square summability for one nice orthonormal basis whereas for $Te_k\to 0$ we need to check for every orthonormal basis which in practice is not feasible whereas testing the Hilbert-Schdmit criterion on one nice orthonormal sequence is quite practical.  


**Redo of Theorem 4.31** 

Let $T\in K\big(X\big)$, for some Banach space $X$, and consider non-zero $\lambda \in \sigma\big(T\big)$. Then $\lambda$ is isolated (recall e.g. principle of isolated zeros in complex analysis).   Via re-scaling, i.e. $\lambda^{-1}\big(T-\lambda I\big)= \lambda^{-1}T-I=S-I$ it suffices to consider the case where WLOG $\lambda =1$.  

Suppose for *contradiction* that $\lambda$ is not isolated.  This means the intersection of spectrum and punctured balls $\sigma\big(T\big)\cap B^*\big(1,2^{-k}\big)\neq \emptyset$ for all $k\in \mathbb N$ hence there is a sequence of distinct points $\sigma\big(T\big)\ni\lambda_k\to 1$.  By Theorem 4.29 we know $\text{image }\big(T-\lambda_k I\big)$ is closed for all $k\in \mathbb N$ hence the $\lambda_k$ cannot be continuous spectrum [p. 119 from chp 5].  Thus each $\lambda_k$ is either point (eigenvalue) spectrum or residual spectrum, which in in this case means   

$\text{image }\big(T-\lambda_k I\big)=\overline{\text{image }\big(T-\lambda_k I\big)}\subsetneq X$
And by ex 3.8 [Hahn-Banach corollary 3.6] this means $\Big(\text{image }\big(T-\lambda_k I\big)\Big)^\perp \neq \big\{0\big\}$ i.e. the annihilator is non-trivial.  Applying ex 4.8.23 from Eidelman et al we have $\ker \big(T^*-\lambda_k I\big)=\ker \big((T-\lambda_k I)^*\big) =\Big(\text{image }\big(T-\lambda_k I\big)\Big)^\perp \neq \big\{0\big\}$, i.e. $\lambda_k$ is an eigenvalue of $T^*$ when $\lambda_k$ is residual spectrum of $T$.   

Thus there must be a countably infinite number of distinct eigenvalues in $B\big(1,\frac{1}{2}\big)$ for at least one of $T$ or $T^*$ where we also know $T^*$ is compact [proven in Eidelman et al Theorem 4.4.1 though MacCluer proves a special case of this in Prop 4.12].  Via re-labelling if needed e.g. of operators $A:=T^*$ and possibly of subsequences, we need only consider the case that $T$ has a countably infinite number of distinct eigenvalues $\lambda_k\in B\big(1,\frac{1}{2}\big)$.  

The finish is essentially the same argument that I gave for the general argument to ex 4.29  

$W_m:= \ker\big(T-\lambda_1\big)\oplus \dotso\oplus \ker\big(T-\lambda_m\big)$  
Each $W_m$ is finite dimensional [since each kernel is finite dimensional by ex 4.10(c) / ex 4.25] hence $W_m$ is closed [ex 4.4 or as a corollary to ex 3.8]. 

Set $M_1:=\text{span }\big\{v_1\big\}$ where $v_1$ is a unit length vector in $W_1$.   

By Riesz Lemma [a corollary to my answer in ex 3.8] we have  
$M_2:=\text{span }\big\{v_1,v_2\big\}$ where $\big\Vert v_2\big\Vert =1$ with $v_2 \in W_2$ and $\text{dist }\big(v_2,M_1\big)\gt \frac{1}{2}$.  And inductively we have  
$M_r:=\text{span }\big\{v_1,v_2,\dotso, v_r \big\}$ where $v_r\in M_r$ and $\text{dist }\big(v_r,M_{r-1}\big)\gt \frac{1}{2}$

What this means is we may write $v_r =\sum_{k=1}^n w_k$ where $w_k$ observes $Tw_k = \lambda_k \cdot w_k$ though $w_k$ may be the zero vector, and  
$\text{dist }\big(w_r,M_{r-1}\big)=\big\Vert w_r\big\Vert_{X/W_{r-1}} =\big\Vert  w_r+ \sum_{k=1}^{r-1}  w_k\big\Vert_{X/W_{r-1}}=\text{dist }\big(w_r + \sum_{k=1}^{r-1}  w_k,M_{r-1}\big)=\text{dist }\big(v_r,M_{r-1}\big)\gt \frac{1}{2}$
since $ \sum_{k=1}^{r-1}  w_k\in M_{r-1}$ 

Thus $\big\{v_k\big\}_{k=1}^\infty$ is a bounded sequence in $X$ but $\big\{Tv_k\big\}_{k=1}^\infty$ has no Cauchy subsequence which contradicts $T$ being compact.  Note this holds since for any $k\in \mathbb N$ and any $r\gt k$ we have  
$\big\Vert Tv_r -Tv_k\big\Vert = \text{dist }\big(Tv_r, \big\{Tv_k\big\}\big)\quad\quad\text{( and since }\big\{Tv_k\big\}\subseteq W_{r-1}\text{)}$  
$\geq \text{dist }\big(Tv_r, W_{r-1}\big)=\text{dist }\big(\lambda_r w_r +\sum_{k=1}^{r-1} \lambda_k v_k, W_{r-1}\big)=\text{dist }\big(\lambda_r w_r , W_{r-1}\big) =\big\Vert \lambda_r \cdot w_r\big\Vert_{X/W_{r-1}} = \vert\lambda_r\vert \cdot\big\Vert w_r\big\Vert_{X/W_{r-1}}$  
$\geq \frac{1}{2}\cdot\big\Vert w_r\big\Vert_{X/W_{r-1}}$  
$\geq \frac{1}{4}$

Non-zero spectrum of $T$ is isolated.  

**Corollary:**  
If we look at any annulus bounded away from zero, i.e. $A=\big\{z: \delta \lt \vert z\vert \lt \Vert T\Vert \big\}$ for any $\delta \gt 0$ then $\overline A$ is compact hence $\big\vert \sigma\big(T\big)\cap \overline A\big\vert \lt \infty$, i.e. it is a finite set by the same argument as in complex analysis [Beardon]-- do any open covering $\overline A\subseteq \bigcup_{z \in \overline A} B\big(z, \delta^{(z)}\big)$ where the punctured ball $B^*\big(z, \delta^{(z)}\big)$ has no spectrum of $T$  in it, since non-zero spectrum is isolated, and by compactness, there is a finite sub-covering, i.e. $\overline A \subseteq \bigcup_{k=1}^m B_k$ where the balls $B_k$ have at most one point of spectrum (the center).  [Note: as we see in chapter 5 via Neumann series argument, any point with modulus $\gt \Vert T\Vert$ is a regular point for any bounded operator on a Banach space].  

**Corollary: all non-zero spectra are eigenvalues (option 1)**   
All non-zero spectrum of a compact operator is necessarily an eigenvalue.  Using ex 5.5, if $0\neq \lambda\in \sigma\big(T\big)$ then $\lambda$ is isolated so $\lambda \in \partial\sigma\big(T\big)$ hence $\lambda$ is an *approximate eigenvalue*.  Now if $\big(T-\lambda I\big)$ is injective, then it is bounded below since its image is closed [Theorem 4.29 then ex 3.27] so $\lambda$ cannot be approximate point spectrum [ref the definition in ex 5.5] hence $\big(T-\lambda I\big)$ being injective implies $\lambda$ is a regular point.  Conclude all non-zero spectra are necessarily eigenvalues for $T$.  

**Corollary: all non-zero spectra are eigenvalues (option 2)**   
By using holomorphic functional calculus we can get this result without even using Theorem 4.29.  Select and fix some $\lambda \in \sigma\big(T\big)\setminus\big\{0\big\}$. Similar to ex 6.19, Let $f$ be a function holomorphic on a (not necessarily connected) open set containing $\sigma(T)$ such that $f\big(\lambda\big)=1$ and $f\Big(\sigma\big(T\big)\setminus\big\{\lambda\big\}\Big)=\big\{01\big\}$, then $f\big(T\big)$ is a Riesz Projector-- it is idempotent since $f\big(T\big)\cdot f\big(T\big)= \big(f\cdot f\big)\big(T\big)=f\big(T\big)$ and it is bounded since the holomorphic functional calculus sends bounded operators to bounded operators (or e.g. $f(T)$ may be uniformly approximately by a rational function of $T$ which are bounded, etc.).  

$T$ commutes with $f(T)$ so $W:=\text{image }f(T)$ is a $T$-invariant subspace (which is non-trivial since $\sigma\big(f(T)\big)=\big\{0,1\big\}$ by spectral mapping theorem for holomorphic functional calculus so $f(T)$ is a non-trivial projection).  In an argument very similar to ex 5.18, we have $\sigma\big(T\vert_W\big)\subseteq \sigma\big(T\big)$ since if $\zeta\in \sigma(T)^C$, i.e. if $\zeta$ is a regular point of $T$, then $A:=T-\zeta$ is invertible and commutes with $f(T)$ and so does its inverse [an Artin chp 2 group theory problem $A^{-1}f(T)=A^{-1}f(T)AA^{-1}=A^{-1}Af(T)A^{-1}=f(T)A^{-1}$] so $A\vert_W$ has inverse $A\vert_W^{-1}$ and $\sigma\big(T\vert_W\big)^C\supseteq \sigma\big(T\big)^C$. Since via Runge [ref my ex 10.9.15 Eidelman et al writeup] we have $r_k(T)\to T$ uniformly where $\big\{r_k\big\}$ is a sequence of rational functions, and since necessarily e.g. $\big\Vert T\vert_W\big\Vert \leq \big\Vert T\big\Vert$, we can confirm that $f\big(T\vert_W\big) = \text{id}_W$ by considering strong convergence [recall e.g. ex 2.23], i.e. if $w\in W$ then $f\big(T\vert_W\big)w = \lim_{k\to\infty}r_k\big(T\vert_W\big)w = \lim_{k\to\infty} r_k\big(T\big)w= f\big(T\big)w=w$.   
$\implies \sigma\big(T\vert_W\big)=\big\{\lambda\big\}$ by spectral mapping theorem for holomorphic functional calculus since if  
$\sigma\big(T\vert_W\big)\neq \big\{\lambda\big\}$ then $0\in f\Big(\sigma\big(T\vert_W\big)\Big)=\sigma\Big(f\big(T\vert_W\big)\Big) = \sigma\big(\text{id}_W\big)=\big\{1\big\}$ which is absurd.  

Thus $\sigma\big(T\vert_W\big)=\big\{\lambda\big\}$ so $T\vert_W$ is compact [ex 4.21] and invertible so $W$ is finite dimensional [ex 4.24] and thus $\lambda$ is an eigenvalue by finite dimensional linear algebra.  

Note that the closure of the image of a compact operator $T\in B(\mathcal H)$ is a separable Hilbert space since  
$\overline{\text{image }T}\subseteq \mathcal H$ is closed hence complete and it inherits the inner product from $\mathcal H$. Further  
$\overline{\text{image }TT^* } = \ker \big(TT^*\big)^\perp = \big(\ker T^*\big)^\perp = \overline{\text{image }T}$  
and we have a countable orthonormal basis for $\overline{\text{image }TT^*}$ given by the eigenvectors of with positive eigenvalues of $TT^*$ [spectral theorem for self-adjoint compact operators].  

**A broken problem / theorem in Kreyszig**  
i.e. problem 9.11.9

> A bounded self-adjoint linear operator $T:H\longrightarrow H$ on a Hilbert space $H\neq \big\{0\big\}$ is said to have a purely continuous spectrum if $T$ has no eigenvalues.  If $T\in B(H)$ is self-adjoint, show that there is a closed subspace $Y\subseteq H$ which reduces $T$ and is such that $T\vert_Y$ has pure point spectrum and $T\vert_{Y^\perp}$ has a purely continuous spectrum.  (This reduction facilitates the investigation of $T$.)    

*Purported solution from back of the Kreyszig*    
> Let $Y$ be the closure of the span of all eigenvectors of $T$.  Then $T_1:=T\vert_Y$ has pure point spectrum and $T_1Y\subseteq Y$.  Also $T_1$ is self-adjoint on $Y$.  Similarly $T_2=T\vert_{Y^\perp}$ is self adjoint on $Y^\perp$ and has a purely continuous spectrum as follows from construction of $Y$.  


This problem immediately struck me as nonsensical due to the implications it would have for injective self-adjoint compact operators.  The compact operators counter-argument would be e.g. let $H=\mathscr l^2$ and $Te_k = \frac{1}{k}\cdot e_k$.  Then $T$ is self-adjoint [e.g. by ex 4.12] and compact [in fact Hilbert-Schmidt]. If $v\in \ker T$ then $0 = \big\langle Tv, e_k\big\rangle = \big\langle v, Te_k\big\rangle = \frac{1}{k}\big\langle v, e_k\big\rangle\implies v =\mathbf 0$ since $v\perp\text{span }\big\{e_k\big\}$ which is dense in $\mathscr l^2$ [or e.g. Theorem 1.33(b)] so $T$ is injective.  Let $Y$ exist as in the 'solution' as a closed subspace and $T\vert_Y$ is thus compact [ex 4.21], injective, and has pure point spectrum hence $0$ is a regular point (since it is not an eigenvalue) and we conclude $T\vert_Y$ is a compact invertible operator with infinite dimensional image which is absurd.

A non-compact operator counter-argument (using chapter 5 ideas):  
Let $H=\mathscr l^2$ and let $\big\{r_k\big\}$ be a counting of rationals in $\big[0,1\big]$, with $Te_k = r_k\cdot e_k$.  Then $Y=\mathscr l^2$ since $\text{span}\big\{e_k\big\}$ is dense in $\mathscr l^2$ and $T\vert_Y=T$ has only point spectrum, which includes all the rationals in $\big[0,1\big]$ but $\sigma\big(T\big)=\big[0,1\big]$ since the spectrum is closed.  Eigenvectors with different eigenvalues for a self-adjoint operators are orthogonal [ex 4.11(b)] hence $\mathscr l^2$ (which is separable) has an uncountable number of orthonormal vectors and this is absurd; for a more explicit finish consider $\lambda =\frac{\sqrt{2}}{2}\in \big[0,1\big]\setminus \mathbb Q$ so it has an associated eigenvector $v$ that satisfies $\lambda \cdot \big\langle v, e_k\big\rangle= \big\langle T\vert_Y v, e_k\big\rangle =\big\langle  v, T\vert_Y e_k\big\rangle=r_k\cdot \big\langle v, e_k\big\rangle\implies \big\langle v, e_k\big\rangle=0$ for all $k\in \mathbb N\implies v=\mathbf 0$ which is absurd.  

In either case, what this shows is that when $T$ has countably infinite number of orthonormal (eigen)vectors, then the closure of the span of those eigenvectors can force continuous spectrum to 'creep in'.  And of course recalling ex 3.12, if we just look at the span of a countably inifinite number orthonormal (eigen)vectors then that space cannot be complete (i.e. Hilbert).

Let $X$ be a $n-$dimensional normed linear space over $\mathbb C$.  Then there is a linear homeomorphism $T:X\longrightarrow \mathbb C^n$.  For convenience we may equip $\mathbb C^n$ with the standard 2-norm, though we know from Theorem 4.2, or my corollaries to ex 3.8, that any choice of norm will be topologically equivalent.

**4.3**  
Suppose that $X$ is a normed linear space, endowed with the metric topology, and suppose $X$ contains a nonempty open set $V$ such that $\overline V$ is compact. The goal of this problem is to show that this forces $X$ to be finite-dimensional.  

**note:** $V$ is necessarily **bounded** since $V\subseteq \overline V$ which is bounded by compactness --i.e. every sequence in $\overline V$ is has a convergent subsequence so $\overline V$ cannot have unbounded sequences hence cannot be unbounded.    

**(a.)** Without loss of generality we may assume that $\mathbf 0\in V $ [translation is a homeomorphism].  Show that as $x$ ranges over the set $V$, the open sets  
$x+\frac{1}{2}V:= \big\{x+\frac{1}{2}v:v\in V\big\}$ form an open cover of $\overline V$.  

Note that scaling by a non-zero number [as well as translation] is a homeomorphism so $x+\frac{1}{2}V$ is open.  

(i.) if $y \in V$ then let $x:=y\implies  x+\frac{1}{2}V= y+\frac{1}{2}V\ni \big(y+\frac{1}{2}\cdot \mathbf 0\big)=y$     

(ii.) if $y \in \overline V\setminus V$ then there exists $ V\ni y_k\to y$ [metric spaces are second countable] and since $\mathbf 0\in\frac{1}{2}V$ which is open, then this means there is some $\epsilon \gt 0$ such that $B\big(\mathbf 0,\epsilon\big)\subseteq \frac{1}{2}V$, so select $k$ large enough such that $\big\Vert y_k-y\big\Vert \lt \epsilon$, i.e. $y\in B\big(y_k, \epsilon\big)\subseteq \big(y_k  +\frac{1}{2}V\big)$       

Conclude: $V\subseteq \bigcup_{x \in V}\big(x+ \frac{1}{2}V\big)$  

By compactness of $\overline V$ there is a a finite subcover, $\big\{x_k +\frac{1}{2}V\big\}_{k=1}^N$  for some $N\in \mathbb N$.  
$Y:= \text{span }\big\{x_1,x_2,\dotso, x_N\big\}$.  

**(b.)**  Show that $V\subseteq Y+ \frac{1}{2^m}V$ for $m\in \mathbb N$ hence  $V\subseteq \bigcap_{m=1}^\infty \big(Y+\frac{1}{2^m}V\big)$  
Formally proceed by induction, where we have the Base Case: $V\subseteq Y+ \frac{1}{2}V$.  

Inductive case: 
Select integer $m\geq 2$.  By inductive hypothesis we know  
$V\subseteq Y+ \frac{1}{2^{m-1}}V$  
$\implies \frac{1}{2}V\subseteq \frac{1}{2}\big(Y+ \frac{1}{2^{m-1}}V\big)=\frac{1}{2}Y+ \frac{1}{2^m}V=Y+ \frac{1}{2^m}V\quad\text{(Y is a vector space so closed under scaling)}$  
$\implies Y+\frac{1}{2}V\subseteq Y + \big(Y + \frac{1}{2^{m}}V\big)=\big(Y + Y\big) + \frac{1}{2^{m}}V = Y+ \frac{1}{2^m}V\quad\text{(Y is a vector space so closed under addition)}$   
$\implies V\subseteq Y+\frac{1}{2}V\subseteq Y+ \frac{1}{2^m}V\quad\text{(Base Case combined with prior line)}$   
$\implies V\subseteq \bigcap_{m=1}^\infty \big(Y+\frac{1}{2^m}V\big)\quad\text{(since it holds for each set in the descending filtration)}  $  


**(c.)**  Show that $Y=\bigcap_{m=1}^\infty \big(Y+\frac{1}{2^m}V\big)$  
By definition $p\in Y\implies p \in \bigcap_{m=1}^\infty \big(Y+\frac{1}{2^m}V\big)$  

$p \not \in Y\implies p\not \in \bigcap_{m=1}^\infty \big(Y+\frac{1}{2^m}V\big)$  
Since $V$ is bounded, i.e. for $v\in V$ then $\big\Vert v\big\Vert \leq C\lt\infty\implies $ for $v\in \frac{1}{2^m} V$ then $\big\Vert v\big\Vert \leq \frac{C}{2^m}$           


Now suppose $p \not \in Y$, which is closed since it is a finite dimensional vector space [ref e.g, ex 4.4 or my corollary to ex 3.8], hence $\text{dist }\big(p, Y\big) =d \gt 0$ [recall McCleary ex 3.6].  Using Hahn-Banach Corollary 3.6 [ex 3.8] there is some $\varphi \in X^*$ with $Y\in \ker \varphi$ and $\varphi\big(p\big)=d$ and $\big\Vert \varphi\big\Vert =1$ hence for $m$ large enough (in particular we want to select $m$ large enough such that $d \gt \frac{C}{2^{m}}$)   we can deduce that  $p\neq  y+ v$ for any choice of $y\in Y$ and $ v\in \frac{1}{2^m}V$ since by reverse triangle inequality and the definition of the norm of $\varphi$     

$\Big\vert \varphi\Big(p- \big(y+v\big)\Big)\Big\vert =\Big\vert \varphi\Big(p\Big)- \varphi\Big(y\Big)-\varphi\Big(v\Big)\Big\vert=\Big\vert d -0-\varphi\Big(v\Big)\Big\vert \geq d -  \Big\vert \varphi\Big(v\Big)\Big\vert\geq d-\big\Vert \varphi\big\Vert \cdot \big\Vert v\big\Vert =d -  1\cdot \big\Vert v\big\Vert\geq d -\frac{C}{2^{m}}\gt 0$  

Conclude $ p\not \in \big(Y+\frac{1}{2^m}V\big)$ for $m$ large enough hence $p\not \in \bigcap_{m=1}^\infty \big(Y+\frac{1}{2^m}V\big)$  

*Note:* since we are using the vector space addition structure for the "+" in $Y+\frac{1}{2^m}V$, using a linear functional from Hahn-Banach was convenient.  However if we did not want to use Hahn-Banach we could make the argument directly using norms, i.e. as above consider  
$\big\Vert p -(y+v)\big\Vert= \big\Vert (p -y) -v\big\Vert \geq \big\Vert p -y\big\Vert  -\big\Vert v\big\Vert \geq \text{dist }\big(p, Y\big)  -\big\Vert v\big\Vert = d  -\big\Vert v\big\Vert \geq d  -\frac{C}{2^m}\gt 0$

**(a.)** Suppose $X$ is a normed space with closed subspace $M$ and finite dimensional subspace $V$.  Show that $M+V$ is closed.  


It is enough to consider the case where $M+V=M\oplus V$.  In particular induct on the dimension and use bunching.  If $M+V\neq M\oplus V$ then we can find a non-zero vector $v$ in common and write this as $M+V'$ where $V'$ has dimension one less [i.e. make a basis for $V$ starting with $v$ then $V'$ has that same basis with $v$ removed]. 
Now, we have a Base Case where $\dim V=1$ and prove $M\oplus V$ is closed; note there is nothing to do when $\dim V=0$. For the inductive case, where $\dim V\geq 2$ via bunching, we have $M\oplus V = \big(M \oplus \text{span}\{v'\}\big)\oplus V'$ where $M\oplus\text{span}\big\{v'\big\}$ is closed by the $1-$dimensional Base Case and $\dim V'=\dim V-1$ so the result follows by  induction hypothesis.  I.e. this is one of those less common inductive problems where the inductive case is trivial and the Base Case takes some work.  

*Base Case:* $M\oplus V$ is closed when $\dim V=1$   
Consider arbitrary sequence $\big\{m_k + \alpha_k\cdot x_0\big\}$, with $m_k \in M$ and $V=\text{span }\big\{x_0\big\}$ such that $(m_k + \alpha_k\cdot x_0)\to y\in X$.  It is convergent hence Cauchy.  
Let $\varphi: M\oplus \text{span}\big\{x_0\big\}\longrightarrow \mathbb C$ be given by $\varphi\big(M\big)=\big\{\mathbf 0\big\}$ and $\varphi\big(x_0\big)=1$.   
$\varphi$ is a linear functional by construction and it is continuous/bounded since its kernel is closed [ref 'optional corollary: all linear functionals on a finite dim vector space are continuous' after ex 3.8 in 'MacCluer_chp3.ipynb'-- the *lemma* there gives a simple quotient argument for this result that does not introduce any circularity with this exercise and does not depend on Hahn-Banach].  


Thus for any $\epsilon \gt 0$, for $j,k\geq K$ we have  
$\big\vert\alpha_k-\alpha_j\big\vert=\big\vert \varphi\big(m_k + \alpha_k\cdot x_0-(m_j + \alpha_j\cdot x_0)\big)\big\vert \leq \big\Vert \varphi\big\Vert \cdot \big\Vert m_k + \alpha_k\cdot x_0-(m_j + \alpha_j\cdot x_0)\big\Vert\lt \big\Vert \varphi\big\Vert \cdot\epsilon$    
$\implies \big\{\alpha_k\big\}$ is Cauchy hence $\alpha_k\to \alpha_0$ since $\mathbb C$ is complete. [Note: I do not use bounded sequences of complex numbers having convergent subsequences like in the problem's hint.]  
$\implies M\ni \lim_{k\to\infty} m_k = \lim_{k\to\infty}\Big(  \big(m_k + \alpha_k\cdot x_0 \big)-\alpha_k\cdot x_0\Big)= \lim_{k\to\infty}  \big(m_k + \alpha_k\cdot x_0 \big)-\lim_{k\to\infty}\alpha_k\cdot x_0 = \big(y-\alpha_0\cdot x_0\big)$    
since all $m_k\in M$ and $M$ is closed  
$\implies \lim_{k\to\infty}\big(m_k + \alpha_k\cdot x_0 \big)=y = \big(y-\alpha_0\cdot x_0\big) + \alpha_0\cdot x_0\in M\oplus \text{span}\big\{x_0\big\}$  
$\implies M\oplus \text{span}\big\{x_0\big\}$ is a closed subspace of $X$.

**extension**  
Using a mixture of the above Banach space arguments and Hilbert space arguments [this is closely related to exercise 4 and 5 in section 10A of Axler's *MIRA*]  

Every finite rank operator $T\in B(\mathcal H, \mathcal H')$ can be described as  
$Th = \sum_{k=1}^n \big\langle h,x_k\big\rangle\cdot y_k= \sum_{k=1}^n T_k h$  where each $T_kh= \big\langle h,x_k\big\rangle\cdot y_k$ [rank one]   

The adjoint of $T_k$ is given by, for each $h \in \mathcal H$ and $w\in \mathcal H'$    
$\big\langle h,x_k\big\rangle\cdot\big\langle  y_k ,w\big\rangle=\Big\langle \big\langle h,x_k\big\rangle\cdot y_k ,w\Big\rangle = \Big\langle T_kh,w\Big\rangle =  \Big\langle h,T_k^*w\Big\rangle =  \Big\langle h,\big\langle w,y_k\big\rangle \cdot x_k\Big\rangle =  \overline{\big\langle w,y_k\big\rangle} \cdot\big\langle h, x_k\big\rangle=  \big\langle y_k,w\big\rangle \cdot\big\langle h, x_k\big\rangle$  
$\implies T_k^*w = \big\langle w,y_k\big\rangle \cdot x_k$  

and $T^* = \sum_{k=1}^n T_k^*\implies T^*= \sum_{k=1}^n \big\langle w,y_k\big\rangle \cdot x_k$   
i.e. this directly shows that $T$ has finite rank *iff* $T^*$ does [the other direction of the iff holds since $T^{**}=T$]

However this need not be true with respect to kernels.  Consider e.g. $T\in B\big(\mathscr l^2\big)$ where

$T^*$ sends even indices to half their index position and $T^*$ kills odd indices so $\dim \ker T=0$ but $\dim \ker T^*=\infty$

**(c.)**   Show that if $T$ is compact in $B(\mathscr H)$ and $\lambda \neq 0$ then $\ker T-\lambda I$ is finite dimensional.  $W:=\ker T-\lambda I$ is a closed $T-$invariant subspace and hence $T\vert_W$ is compact [ex 4.21].  and $T\vert_W=\lambda \cdot \text{id}$, which is invertible hence $\dim W \lt \infty$ per part (b.).  
Note: as with part (b.) and ex 4.21, this argument runs identically on a Banach space.

**(a.)**  
Suppose $T$ is a normal operator on $B(\mathscr H)$.  

$\big\Vert (T-\lambda I)h\big\Vert^2 =\big\langle (T-\lambda I)h,(T-\lambda I)h\big\rangle = \big\langle (T-\lambda I)^*(T-\lambda I)h,h\big\rangle= \big\langle (T-\lambda I)^*h,(T-\lambda I)^*h\big\rangle=\big\Vert (T-\lambda I)^*h\big\Vert^2$  
$\implies \ker \big(T-\lambda I\big)=\ker \big(T-\lambda I\big)^*$  

Note this means $\big(T -\lambda I\big)v=\mathbf 0\iff \big(T^* -\overline \lambda I\big)v=\mathbf 0$  or equivalently $Tv=\lambda \cdot v\iff T^*v =\overline\lambda v$  

Also, since $\overline{\text{image T}}=\ker (T^*)^\perp=\big(\ker T\big)^\perp$  
[ref ex 2.16 and ex 1.12] this means  
$\mathscr H = \overline{\text{image T}}\oplus \ker T$  


**(b.)**   
Show that if $T$ is normal then eigenvectors corresponding to distinct eigenvalues are orthogonal  
let $Tv =\lambda \cdot v$ and $Th= \lambda'\cdot h$  for $\lambda \neq \lambda '$  

$\lambda'\cdot \big\langle h,v\big\rangle= \big\langle Th,v\big\rangle = \big\langle h,T^*v\big\rangle = \big\langle h,\overline \lambda \cdot v\big\rangle=\lambda \cdot \big\langle h, v\big\rangle $  
$\implies \big(\lambda'-\lambda\big)\cdot \big\langle h,v\big\rangle=0\implies h\perp v$  

**corollary:**  
A normal operator on a separable Hilbert space can only have countably many distinct eigenvalues [since if it had uncountably many distinct eigenvalues then it would have an uncountable orthonormal set hence not be separable].   

Note if a Hilbert space has an uncountable orthonormal set $\big\{e_\alpha\big\}$ then it cannot be separable. Suppose not: then by separability there is a countable set $\big\{s_k\big\}_{k=1}^\infty$ where for say $\delta=\frac{1}{2}$ we have, for any given $\alpha$ there is some $k$ such that $\big\Vert e_\alpha - s_k\big\Vert \lt \frac{1}{2}$  but for $\beta \neq \alpha$ then it is impossible for  
$B\big(e_\alpha,\frac{1}{2}\big)\ni s_k\in B\big(e_\beta,\frac{1}{2}\big) $ since that would imply  
$\sqrt {2}=\big\Vert e_\alpha -e_\beta \big\Vert \leq \big\Vert e_\alpha -s_k\big\Vert +\big\Vert s_k -e_\beta \big\Vert\leq 1$ which is absurd.  Thus for each $\alpha$ in our uncountable set we assign a distinct $s_k$ to $B\big(e_\alpha,\frac{1}{2}\big)$ [i.e. separability here implies an injective mapping from $\big\{e_\alpha\big\}$ to $\big\{s_k\big\}_{k=1}^\infty$] so $\big\{s_k\big\}_{k=1}^\infty$ is uncountable which is absurd.  

**(c.)**  
State and prove a version of Theorem 4.21 for compact normal operators  
If $T$ is a compact normal operator in $B(\mathscr H)$ then the set of eigenvalues of $T$ is countable and if not finite then the modulus tends to zero.  In fact we can strengthen this statement quite a bit by adapting Theorem 4.24, compact spectral theorem for self-adjoint operators, to that for compact normal operators. That will show that $T$ has countably many distinct non-zero eigenvalues, whose modulus tends to zero if an infinite set, plus perhaps the eigenvalue $0$. 

We need only address the case where the image of $T$ has a countably infinite orthonormal basis. For the finite rank case, we can use reduce the result to that of regular finite dimensional linear algebra by writing  $\mathscr H = \overline{\text{image T}}\oplus \ker T^*= \overline{\text{image T}}\oplus \ker T=\text{image T}\oplus \ker T$ where the 1st equality is ex 4.11(a), and of course finite dimensional subspaces are closed; then use the fact that each are $T$ invariant hence $T\vert_\text{image T}$ is a finite dimensional linear operator [ref commentary under 'notes on the square-root under ex 3.37 in 'MacCluer_chp3.ipynb' for further commentary along these lines].  

For arbitrary $\sigma \in \mathbb R_{\gt 0}$, set $M:= \ker \big(T^*T-\sigma I\big)$ where $\dim M\lt \infty $ per ex 4.10(b).  Using the operator matrix formalism of ex 4.18, we have [ref 'eidelman_et_al_FunctionalAnalysisAnIntroduction_chp6.ipynb' for additional commentary on operator matrices]   

$T=\left[\begin{matrix}T\vert_M & \mathbf 0 \\ \mathbf 0 & T\vert_{M^\perp} \end{matrix}\right]$   
as $M$ is $T$ and $T^*$ invariant. 
$\implies T=\left[\begin{matrix}T\vert_M^* & \mathbf 0 \\ \mathbf 0 & T\vert_{M^\perp}^* \end{matrix}\right]$ and in particular  $T\vert_M^*T\vert_M=T\vert_MT\vert_M^*$ i.e. the restriction is normal and finite dimensional hence has an orthonormal basis of eigenvectors and eigenvalues with squared modulus of $\sigma$ per the spectral theorem for finite dimensional Linear Algebra.    

Theorem 4.24 tells us that if $v\in \text{image }T^*T$ then $v=T^*Tx =\sum_{k=1}^\infty \sigma_k \big\langle x,g_k\big\rangle \cdot g_k$ 
where the $g_k$ are an orthonormal set.  What the above invariant subspace argument tells us is we may *choose* this orthonormal set to be eigenvectors of $T$.  E.g. if $\dim\ker \big(T^*T-\sigma_1 I\big)= m$ we may map $\big\{g_1,\dotso,g_m\big\}\mapsto \big\{u_1,\dotso,u_m\big\}$ where each set is orthonormal and $\text{span }\big\{g_1,\dotso,g_m\big\}= \text{span }\big\{u_1,\dotso,u_m\big\}$ and the $u_k$ are eigenvectors of $T$.  Thus  

$v\in \text{image }T^*T$ then $v=T^*Tx =\sum_{k=1}^\infty \sigma_k \big\langle x,u_k\big\rangle \cdot u_k\subseteq \overline{\text{span }\big\{u_1,u_2, \dotso\big\}}\quad\text{(for some }x\in\mathscr H\text{)}$ 

$\implies \text{image }T\subseteq \overline{\text{image }T} = \big(\ker T\big)^\perp = \big(\ker T^*T\big)^\perp= \overline{\text{image }T^*T}\subseteq \overline{\text{span }\big\{u_1,u_2, \dotso\big\}}$  
(in fact the RHS should be an equality but we don't need that)  
Recall ex 2.16 for the 2nd equality.  The fact that $\ker T^*T=\ker T$ is true because if $x\in \ker T$ then $T^*\big(Tx\big) =\mathbf 0\implies x \in \ker T^*T$.  Conversely if $x\in \ker T^*T$ then   $0=\big\langle  T^*Tx, x\big\rangle =\big\langle  Tx, Tx\big\rangle =\big\Vert Tx\big\Vert^2\implies x\in \ker T $  


Conclude: $v\in  \text{image }T\implies Tx =v=\sum_{k=1}^\infty \big\langle v,u_k\big\rangle\cdot u_k=\sum_{k=1}^\infty \big\langle Tx,u_k\big\rangle\cdot u_k=\sum_{k=1}^\infty \big\langle x,T^*u_k\big\rangle\cdot u_k=\sum_{k=1}^\infty \lambda_k\cdot\big\langle x,u_k\big\rangle\cdot u_k$  
where $\lambda_k = \alpha_k\cdot\sqrt{\sigma_k}$ for some $\alpha_k$ in the unit circle.  Note the 2nd equality in the above line is Theorem 1.33(c).    

Note that with the possible exception of the eigenvalue zero, $T$ cannot have any eigenvalues other than those in $\big\{\lambda_1,\lambda_2,\lambda_3,\dotso\big\}$, since if there was some other $\lambda$ not in that set then $Tv =\lambda v$ for some non-zero $v$ and by part (b.) we have $v\perp u_k$ for all $k\implies v \perp
\overline{\text{span }\big\{u_1,u_2, \dotso\big\}}\implies v\perp \overline{\text{image }T}$ and $\mathscr H =\overline{\text{image }T}\oplus\ker T\implies v \in \ker T\implies \lambda =0$

https://math.stackexchange.com/questions/561636/show-that-a-positive-operator-on-a-complex-hilbert-space-is-self-adjoint/      


Another argument is the criterion forces

where the final implication comes from either   

**(i.)** ex 2.14(b.) since $u(x,y):= \big\langle x,\big(T-T^*\big)y\big\rangle=\big\langle \big(T-T^*\big)^*x,y\big\rangle$ is a bounded sesquilinear form that agrees with the zero form on the diagonal hence is the zero form and $\big(T-T^*\big)^*=\mathbf 0=\big(T-T^*\big)$ is thus the unique operator that satisfies this per Theorem 2.11  

**(ii.)**  $\big\langle h,\big(T-T^*\big)h\big\rangle = 0=\big\langle h,i\big(T-T^*\big)h\big\rangle= \big\langle i\big(T-T^*\big) h,h\big\rangle$ for all $\Vert h\Vert =1$ but $i\big(T-T^*\big)$ is self-adjoint so by Lemma 4.18  
$\big\Vert \big(T-T^*\big) \big\Vert=\big\Vert i\big(T-T^*\big) \big\Vert = \sup_{\Vert h\Vert =1}\big\vert \big\langle  i\big(T-T^*\big)h,h\big\rangle\big\vert =0 $  

*note:*    
note in general  
$w(A)= \sup_{\Vert h\Vert =1}\big\vert \big\langle  Ah,h\big\rangle\big\vert$ is known as the **numerical radius** of an operator on a Hilbert space and it is *equivalent* to the norm of the operator since   
$w(A)= \sup_{\Vert h\Vert =1}\big\vert \big\langle  Ah,h\big\rangle\big\vert\leq \big\Vert A\big\Vert\cdot 1^2$ by Cauchy-Schwarz.  

But write $A= \frac{\big(A+A^*\big)}{2}+ \frac{\big(A-A^*\big)}{2}$ a sum of its self-adjoint and skew parts and we get  
$\big\Vert A\big\Vert\leq  \left\Vert \frac{\big(A+A^*\big)}{2}\right\Vert+ \left\Vert\frac{\big(A-A^*\big)}{2}\right\Vert=w\left( \frac{\big(A+A^*\big)}{2}\right)+w\left( \frac{\big(A-A^*\big)}{2}\right)\leq w\big(A\big)+w\big(A\big)=2 \cdot w\big(A\big)$     
The 1st equality is Lemma 4.18, and observing that $\frac{A-A^*}{2}$ and $i\cdot\frac{A-A^*}{2}$ necessarily have the same norm and numerical radius though the latter is self-adjoint (so Lemma 4.18 applies to it).  The inequality is e.g. because $\frac{1}{2}\big\langle (A+A)^*h, h\big\rangle \in \mathbb R$ and $\frac{1}{2}\big\langle (A-A)^*h, h\big\rangle \in i\cdot\mathbb R$ hence $\big\vert \frac{1}{2}\big\langle (A+A)^*h, h\big\rangle \big\vert\leq \big\vert \frac{1}{2}\big\langle (A+A)^*h, h\big\rangle+\frac{1}{2}\big\langle (A-A)^*h, h\big\rangle \big\vert = \big\langle Ah, h\big\rangle$ for any $\Vert h\Vert =1$ so take the supremums and by essentially the same argument  $\big\vert \frac{1}{2}\big\langle (A-A)^*h, h\big\rangle \big\vert\leq  \big\langle Ah, h\big\rangle$.  

$\implies w\big(A\big)\leq \big\Vert A\big\Vert\leq 2\cdot w\big(A\big)$ which implies positive definiteness of the numerical radius; triangle inequality comes from the fact that two choices are better than one.  


*remark:*  
If we have dense subspace $W\subseteq \mathscr H$ then it suffices to confirm $\big\langle Tw,w\big\rangle \in \mathbb R$ for all $w \in W$ since for $h\in\mathscr H$ we have $w_k\to h$ and 
$\big\vert \text{imaginary }\big\langle Th,h\big\rangle\big\vert\leq \Big\vert\big\langle Tw_k,w_k\big\rangle-\big\langle Th,h\big\rangle \Big\vert= \Big\vert\big\langle T\big(w_k-h\big),\big(w_k-h\big)\big\rangle\Big\vert\leq\big\Vert T\big(w_k-h\big)\big\Vert \cdot \big\Vert w_k-h\big\Vert\leq \big\Vert T\big\Vert \cdot \big\Vert w_k-h\big\Vert^2\to 0 $  



**4.13**  
In the notation of Corollary 4.25 [i.e. compact self adjoint $T$] show that for a given $\mathbf 0\neq h\in \mathscr H$, we can solve the equation $Tf=h$ for $f \iff h\perp \ker T$ and $\sum_{n=1}^\infty \frac{1}{\lambda_n^2}\vert\langle h,e_n\rangle\vert^2 \lt \infty$.  (I assume $h$ is non-zero to avoid triviality.)  



**(i.)**  Suppose $Tf = h$ (i.e. a solution exists)   
$\mathscr H = \ker T\oplus \big(\ker T\big)^\perp$  
Now suppose $h\in \ker T^\perp =\overline{\text{image }T}$ 
we want to find criterion to ensure that $h\in \text{image }T$.  Note that $\text{image }T\subsetneq \overline{\text{image }T}$ when the image is infinite dimensional per ex 4.26.    

$\overline{\text{image }T}$ has a countable orthonormal basis, the eigenvectors of $T$, given by $\big\{e_n\big\}$, with associated eigenvalues $\lambda_n\gt 0$ [note: a minor modification is needed if $\text{rank }T\lt \infty$] so using the projection formula we have  

$h = \sum_{n=1}^\infty \big\langle h, e_n\big\rangle e_n$   
$\mathbf 0 = Tf -h =\sum_{n=1}^\infty \lambda_n \cdot\big\langle f,e_n\big\rangle e_n-\sum_{n=1}^\infty \big\langle h, e_n\big\rangle e_n = \sum_{n=1}^\infty \big\langle (\lambda_n \cdot f - h), e_n\big\rangle e_n$  
$\implies \big(\lambda_n \cdot f - h\big)\perp e_n$ for all $n$   
$\implies \lambda_n\cdot\big\langle  f, e_n\big\rangle = \big\langle  h, e_n\big\rangle $  
*now in general* $f=u+w$ for $u\in \ker T$ and $w\in \ker T^\perp=\overline{ \text{image }T}$ and the former does not impact the solution so for the moment set $w:=f$  
$\implies \big\langle w, e_n\big\rangle =\big\langle f, e_n\big\rangle = \frac{1}{\lambda_n}\big\langle  h, e_n\big\rangle$  

The projection formula [Theorem 1.33(c)] then implies  
$w = \sum_{n=1}^\infty \big\langle w, e_n\big\rangle e_n =\sum_{n=1}^\infty \frac{1}{\lambda_n}\big\langle  h, e_n\big\rangle e_n$  
As an aside, note that we know $\left\Vert \sum_{n=1}^\infty \big\langle  h, e_n\big\rangle e_n\right\Vert \lt \infty$ by  Bessel's Inequality or Parseval [Theorem 1.33(e)] since $w\in \ker T^\perp$ by assumption. But we need to inquire about $\left\Vert \sum_{n=1}^\infty \frac{1}{\lambda_n}\big\langle  h, e_n\big\rangle e_n\right\Vert \lt \infty$ 

By assumption such  a $w$ is a valid solution, so it has finite length [i.e. making the below criterion necessary], i.e.    
$\infty\gt \big\Vert w\big\Vert^2 = \left \Vert \sum_{n=1}^\infty \frac{1}{\lambda_n}\big\langle  h, e_n\big\rangle e_n\right\Vert^2=\sum_{n=1}^\infty \frac{1}{\lambda_n^2}\left\vert\big\langle h,e_n\big\rangle\right\vert^2 $    
Note that the $w$ is the unique solution in $\ker T^\perp$ since $T$ is injective there    

**(ii.)** Suppose $h\perp \ker T$ and $\sum_{n=1}^\infty \frac{1}{\lambda_n^2}\left\vert\big\langle h,e_n\big\rangle\right\vert^2 \lt \infty$

$f:=\sum_{n=1}^\infty \frac{1}{\lambda_n}\big\langle h,e_n\big\rangle e_n$  where $f$ is well defined since the coefficients (with respect to orthonormal basis vectors $e_n$) are square summable (hence the partial sums are Cauchy) and  
$Tf= \sum_{n=1}^\infty \frac{1}{\lambda_n} \big\langle h,e_n\big\rangle Te_n = \sum_{n=1}^\infty  \big\langle h,e_n\big\rangle e_n = h$  
where the final equality is because $h\in \big(\ker T\big)^\perp = \overline{\text{image } T}= \overline{\text{span } \big\{e_n\big\}_{n=1}^\infty }$ so equality holds by the projection formula [Theorem 1.33(c)].  


**(iii.)**  Find all such solutions $f$ under these assumptions.  
using quotient/factor space notation, from part (i.) we deduce
$\implies f = w + \ker T$

**4.15**  

Let $M_x$ be the multiplication operator acting on $L^2([0,1],dx)$ by $M_x(f)=x\cdot f$.  Note that $M_x$ is self-adjoint [ref p. 37].  Show that it has no eigenvalues, but many reducing subspaces.  

If $M_x$ has an eigenvalue $\lambda$, then $M_x v = \lambda \cdot v$ where $v=[f]$, an equivalence class of functions in $L^2([0,1],dx)$, where $f$ is a distinct representative for that class, which implies   
$x\cdot f - \lambda \cdot f=(x-\lambda)\cdot f =0\text{ a.e. (dx)}$  
where $dx=m$ is of course the standard Lebesgue measure.   

But this means either $(x-\lambda)=0$ on a positive measure set [absurd] or $f$ is the zero function almost everywhere, i.e. $v=\mathbf 0$ so $\lambda$ cannot be an eigenvalue after all.  (Put differently if $f\neq 0$ a.e. then restrict to the set $S$ where $m(S)\gt 0$ such that $f\neq 0$ and deduce $x-\lambda =0$ for $x\in S$ which is absurd.)  

With $\tau \in [a,b]$ for $0\leq a\lt b \leq 1$, let $W_\tau$ be the subspace where if $v=[f]\in W_{\tau}$ then for $0\leq x\leq \tau$, $f(x)=0 \text{ a.e. dx}$.  Then $M_x v\in W_\tau\implies M_x W_\tau\subseteq W_\tau$ and since $M_x$ is self-adjoint, this implies $M W_\tau^\perp \subseteq W_\tau^\perp$ by Corollary 4.23; It is convenient to verify this using the operator matrix formalism of ex 4.18 and write  

$A=\left[\begin{matrix}A\vert_{W_\tau} & B \\ \mathbf 0 &A\vert_{W_\tau^\perp} \end{matrix}\right]=\left[\begin{matrix}A\vert_{W_\tau} & \mathbf 0 \\ \mathbf 0 &A\vert_{W_\tau^\perp} \end{matrix}\right] $   
where necessarily $B=\mathbf 0$ since $ A=A^*=\left[\begin{matrix}A\vert_{W_\tau} & B \\ \mathbf 0 &A\vert_{W_\tau^\perp} \end{matrix}\right]^*=\left[\begin{matrix}A\vert_{W_\tau}^* & \mathbf 0 \\ B^* &A\vert_{W_\tau^\perp}^* \end{matrix}\right]$

ref the beginning of my notes to Eidleman et al chapter 6 for full details on the $^*$ manipulation as it pertains to operator matrices  

Strictly speaking we need to also show that $W_\tau$ is closed, so let characteristic function $\mathcal X_\tau$ be $=0$ for $0\leq z\leq \tau$ and $=1$ otherwise and the associated multiplication operator [ref example 2.5] $M_{\mathcal X_\tau}\in B\big(L^2[0,1]\big)$ since $\mathcal X_\tau$ is essentially bounded.  Now notice that $f\in W_\tau\iff f= \mathcal X_\tau\cdot f = M_{\mathcal X_\tau} f\iff  f\in \text{image }M_{\mathcal X_\tau}$.  Note that $M_{\mathcal X_\tau}^2 v = M_{\mathcal X_\tau}v$  since $M_{\mathcal X_\tau}$ is an (orthogonal) projection, thus $\text{image }M_{\mathcal X_\tau}=\ker \big(I-M_{\mathcal X_\tau}\big)$ is closed.  In other words, $M_{\mathcal X_\tau}$ commutes with $M_{\mathcal X_\tau}$ so $\text{image }M_{\mathcal X_\tau}$ is a (closed) invariant subspace.  

A perhaps more typical approach would be to consider $W_\tau \ni f_k \to f$ (in norm).  Per Chebyshev's Inequality and Riesz' theorem [ref Nelson Project 2(b)] there is a subsequence $f_{k_j}\to f$ point-wise a.e.  Now recall that $f_{k_j}$ is technically an equivalence class (equal a.e.), so select arbitrary distinct representative and call it $g_{k_j}$ where $g_{k_j}(z)=0$ for $z \in [0, \tau]$ except on null set $N_{k_j}$.  Then e.g. $\big\{g_{k_j}\big\}$ is zero on $[0,\tau] \setminus \bigcup_{j=1}^\infty N_{k_j}$ and $g_{k_j}\to f$  point-wise except on some null set $N_0\subseteq [0,\tau]$, thus $f(z)=0$ for $z\in [0,\tau] \setminus \big(N_0\cup\bigcup_{j=1}^\infty N_{k_j}\big)$ where $m\big([0,\tau] \setminus \big(N_0\cup\bigcup_{j=1}^\infty N_{k_j}\big)\big) = 0$ so $f$ is zero almost everywhere on $[0,\tau]$ hence $f\in W_\tau$.

**4.16**  
Show that the Volterra operator $V$ of indefinite integration on $L^2\big([0,1],dx\big)$ defined by  
$$Vf(x)= \int_0^x f(t)dt$$    
is compact.   

Note: technically the work for this has already been done by MacCluer since on page 86 via Theorem 4.16 she proves integral operators on separable $L^2(X,\mu)$ are Hilbert-Schmidt and on page 34 she explained how $V$ is an integral operator with kernel $k(x,y)$ [on $L^2(X,\lambda)$ which is separable] equal to the characteristic function of the lower triangle $\big\{(x,y): y\leq x\big\}$. Thus technically there isn't anything required for this exercise, though I give a direct argument below.  

Using the orthonormal basis $\big\{\exp\big(2\pi i n \cdot x\big)\big\}_{n=-\infty}^\infty$ per https://math.stackexchange.com/questions/1346191/orthonormal-basis-for-mathcall20-1  

for $n=0$ we trivially have $V\exp\big(2\pi i 0 \cdot x\big)=x$  
and for $n\neq 0$  
$$V\exp\big(2\pi i n \cdot x\big)= \int_0^x \exp\big(2\pi i n \cdot t\big)dt=\frac{\exp(2\pi i n \cdot x)-1}{2\pi i\cdot n}$$    
(Technically in the above we are working on a distinct representative that is particularly nice, i.e. a continuous function, and doing the indefinite integration there, and the result then holds for all members of the equivalence class)  

$$\implies \big\Vert V\exp\big(2\pi i n \cdot x\big)\big\Vert  =\left \Vert \frac{\exp(2\pi i n \cdot x)-1}{2\pi i\cdot n}\right\Vert=\frac{1}{2\pi \cdot n}\big\Vert \exp(2\pi i n \cdot x)-1\big\Vert \leq \frac{1}{2\pi \cdot n}\big\Vert \exp(2\pi i n \cdot x)\Vert +\big\Vert 1\big\Vert\leq \frac{2}{2\pi \cdot n}=\frac{1}{\pi \cdot n}$$  
$$\implies \sum_{n=-\infty}^\infty\big\Vert V\exp\big(2\pi i n \cdot x\big)\big\Vert^2=\big\Vert V\exp\big(2\pi i 0 \cdot x\big)\big\Vert^2+\sum_{n=1}^\infty\big\Vert V\exp\big(2\pi i n \cdot x\big)\big\Vert^2+\sum_{n=-1}^{-\infty}\big\Vert V\exp\big(2\pi i n \cdot x\big)\big\Vert^2$$  
$$\leq 1 +2\cdot\sum_{n=1}^\infty\left(\frac{1}{\pi \cdot n}\right)^2= 1+\frac{2}{\pi^2}\sum_{n=1}^\infty \frac{1}{n^2}\lt \infty$$  

Conclude: $V$ is Hilbert-Schmidt hence compact [Theorem 4.15]

**encore:**  

If $\dim M\lt \infty$ and $T$ is normal, then $M$ is $T^*$ invariant (i.e. a reducing subspace).  

*Proof 1:*   
$T^*T =TT^*\implies \begin{bmatrix} X^*X & * \\ * & * \end{bmatrix} = \begin{bmatrix} XX^*+YY^* & * \\ * & * \end{bmatrix}  \implies X^*X =XX^* + YY^*$  
All of these are finite dimensional operators from $M\rightarrow M$, so we may use the trace (including its cyclic property) from linear algebra to conclude  
 $\text{trace}\big(XX^*\big) = \text{trace}\big(X^*X\big) =\text{trace}\big(XX^* + YY^*\big)=\text{trace}\big(XX^*\big) + \text{trace}\big(YY^*\big)=\text{trace}\big(XX^*\big)+\big\Vert Y^*\big\Vert_F^2\implies \mathbf 0 =Y^*$  

and if we examine the blocked structure we have   
$T^*T =TT^*\implies X^*X=XX^*$ and $W^*W=WW^*$    

*Proof 2:*  
If we consider the restriction of $T\vert_M$ and then run Schur Triangularization on it [really: its matrix representation with respect to an orthonormal basis] then we discover it is diagonalizable-- if not then there is some $v$ such that $\big(T-\lambda I\big)v\neq \mathbf 0= \big(T-\lambda I\big)^2 v$.  Consider that $\overline{\text{image} T}=(\ker T)^\perp$ is $\big(T-\lambda I\big)-$invariant by commutativity (or look at the operator matrix implications for $\mathscr H=\ker T \oplus (\ker T)^\perp$) so $w:=\big(T-\lambda I\big) v$  and $w\in (\ker T)^\perp$ but $Tw=\mathbf 0\implies w\in \ker T\cap (\ker T)^\perp\implies w=\mathbf 0$ which is absurd. 
Conclude $M$ has a basis given by (finitely many) eigenvectors of $T$ which must be eigenvectors of $T^*$ by normality, hence $T^*M\subseteq M$.  


**Note:** in the general case of infinite dimensional Hilbert spaces, a normal operator *can* have a block upper triangular structure [i.e. there are counterexamples to $M$ always being reducing] so some form of special structure is needed to ensure a block diagonal operator matrix and $\dim M\lt \infty$ is a particularly nice form of special structure.  


**4.19**  
If $\big\{e_n\big\}_{n=1}^\infty$ is an orthonormal sequence in a Hilbert space $\mathscr H$ and $A\in B(\mathscr H)$ is compact, show that $Te_n\to\mathbf 0$.  

Via considering subsequences if needed, it suffices to suppose for contradiction that there exists an orthonormal sequence $\big\{u_n\big\}$ such that $\big\Vert A u_n\big\Vert \geq 2\epsilon \gt 0$ for all $n$.  With respect to Hilbert spaces, $A\in B(H)$ is compact *iff* it is the limit (in norm) of a sequence of [bounded] finite rank operators $\big\{A_k\big\}$ such that $\big\Vert A_k- A\big\Vert \to 0$, so for $k\geq K$ large enough we have (for all $n$)  
$2\epsilon \leq \big\Vert A u_n\big\Vert= \big\Vert (A-A_k) u_n +A_k u_n\big\Vert\leq \big\Vert (A-A_k) u_n\big\Vert +\big\Vert A_k u_n\big\Vert\leq \epsilon + \big\Vert A_k u_n\big\Vert $  
since $\big\Vert A -A_k \big\Vert\leq \epsilon$ for $k\geq K$ where $\text{rank }A_K =R\lt \infty$   
$\implies \epsilon^2 \leq \big\Vert A_K u_n\big\Vert^2 =\big\langle A_K u_n,A_K u_n\big\rangle$ for all $n$       

Now for $m\gt R=\text{rank }A_K$ consider the $m\times m$  Gram Matrix $G^{(m)}$ with $g_{i,j}^{(m)}:= \big\langle A_Ku_j, A_K u_i\big\rangle$  
$\implies \text{rank }G^{(m)}\leq R$ and with $\lambda_j^{(m)}$ being the eigenvalues of $G^{(m)}$ [in descending order]  
$\implies m\cdot\epsilon^2\leq \text{trace }\big(G^{(m)}\big) = \sum_{j=1}^{R} \lambda_j^{(m)}\leq R \cdot \lambda_1^{(m)}= R\cdot \sup_{\Vert \mathbf q^{(m)}\Vert =1}\big(\mathbf q^{(m)}\big)^* G^{(m)}\mathbf q^{(m)}$  
$ =  R\cdot  \sup_{\Vert \mathbf q^{(m)}\Vert =1}\big\langle A_K\sum_{j=1}^m q_j^{(m)} \cdot u_n, A_K\sum_{j=1}^m q_j^{(m)} \cdot u_n\big\rangle= R\cdot \sup_{\Vert \mathbf q^{(m)}\Vert =1} \big\Vert A_K\sum_{j=1}^m q_j^{(m)} \cdot u_n\big\Vert^2$ $\leq  R\cdot \big\Vert A_K\big\Vert^2$  
where the last inequality holds since $\big\Vert \sum_{j=1}^m q_j^{(m)}\cdot u_n \big\Vert^2 = \sum_{j=1}^m \big\vert q_j^{(m)}\big\vert^2 =1$ (this is where $u_n$ being orthonormal is used)    
Thus $m\cdot\epsilon^2\leq R\cdot\big\Vert A_K\big\Vert^2$ but the upper bound is constant (and finite) while the lower bounded may be made arbitrarily large and that is a contradiction.  

Note: the fact that $\text{rank }G^{(m)}\leq R$, independent of $m$ can be flushed out using Artin chapter 7 [Bilinear Forms] type of arguments.  An expedient argument is to consider $m\gt R$ so $\big\{A_Ku_n\big\}_{n=1}^m$ is a linearly dependent set and $G^{(m)}$ is not invertible per ex 2.20.  This occurs when $m=R+1$ so for $m\geq R+1$ the leading $R+1\times R+1$ principal submatrix is singular and since $G^{(m)}\succeq \mathbf 0$ [recall it can be written as $B_m^*B_m$ per ex 2.20(b.)] this means $G^{(m)}$ has at most $R$ non-zero eigenvalues per Cauchy Interlacing.

**Direct spectral theory argument using Eidelman et al for any Hilbert space (does not require detour into separability)**  
note: the 'second approach to the spectral theorem in Bachman & Narici would work just as well in this argument, with minor modification, as would the approach taken in Riesz & Nagy.    

Using the (Riemann type) spectral integral for bounded self-adjoint operators, i.e. (7.1.8) from Eidelman et al (page 111).  

We have bounded $T\succeq \mathbf 0$ and for simplicity we assume WLOG that $\big\Vert T\big\Vert =1$.  So select $\epsilon \in \big(0,1\big)$ and $n$ large enough such that $\frac{1}{n-1}\lt \epsilon\lt \frac{2}{n-1}$.  Then form the partition, for $k\in \big\{0,1,\dotso, n\big\}$, set $\lambda_k:= \frac{k}{n-1}-\delta$ for $\delta\gt 0$ small enough (which means that $\lambda_1\gt 0$ and $\lambda_2\gt \epsilon$).         

For $1\leq k \leq n$ then $\vert \lambda_k-\lambda_{k-1}\vert =\frac{1}{n-1}\lt \epsilon$.  Set $\mu_0:=0\in \big[\lambda_0, \lambda_1\big]$ and for $k \in \big\{1,\dotso, n\big\}$ set $\mu_k := \lambda_{k+1}\gt \epsilon$.   
The criteria in section 7.1(b) of Eidelman et al are met for constructing the spectral integral so per (7.1.8) we have  

$\big\Vert T-\sum_{k=1}^{n-1}\mu_k \cdot \big(E_{\lambda_{k+1}}-E_{\lambda_k}\big)\big\Vert= \big\Vert T-\sum_{k=0}^{n-1}\mu_k \cdot \big(E_{\lambda_{k+1}}-E_{\lambda_k}\big)\big\Vert \leq \epsilon$  
where the first equality holds since $\mu_0=0$.  Note that each  $\big(E_{\lambda_{k+1}}-E_{\lambda_k}\big)$ is an orthogonal projection [possibly equal to the zero projector] and   
$\big(E_{\lambda_{k+1}}-E_{\lambda_k}\big)\big(E_{\lambda_{m+1}}-E_{\lambda_m}\big)=\mathbf 0$  
for $m\neq k$ [ref e.g. the commentary in the middle of page 112 or work through property (iii) on page 109 and recall property 6.4.8 on page 102] which implies the orthogonal projections commute (take adjoints of the above product) hence $\text{image }\big(E_{\lambda_{k+1}}-E_{\lambda_k}\big)$ is an invariant subspace of $\big(E_{\lambda_{m+1}}-E_{\lambda_m}\big)$ thus $\big(E_{\lambda_{m+1}}-E_{\lambda_m}\big)$ kills everything $\text{image }\big(E_{\lambda_{k+1}}-E_{\lambda_k}\big)$. 

Now suppose for contradiction that $\text{rank }\big(E_{\lambda_{r+1}}-E_{\lambda_r}\big) \not\lt \infty $ for some $r\in \big\{1,2,\dotso, n\big\}$.  Then via Gram-Schmidt we can find an orthnormal sequence $\big\{e_j\big\}\in \text{image }\big(E_{\lambda_{r+1}}-E_{\lambda_r}\big)$  and for $j \in \mathbb N$  
$\epsilon\geq \Big\Vert \Big(T-\sum_{k=1}^{n-1}\mu_k \cdot \big(E_{\lambda_{k+1}}-E_{\lambda_k}\big)\Big)e_j\Big\Vert=\Big\Vert Te_j- \mu_r \cdot \big(E_{\lambda_{r+1}}-E_{\lambda_r}\big)e_j\Big\Vert=\Big\Vert Te_j- \mu_r \cdot e_j\Big\Vert\geq \Big\Vert \mu_r\cdot e_j\Big\Vert - \Big\Vert Te_j\Big\Vert=\mu_r - \Big\Vert Te_j\Big\Vert$  
$\implies \epsilon\geq \mu_r - \lim_{j\to\infty}\Big\Vert Te_j\Big\Vert=\mu_r\gt \epsilon\quad\text{(absurd)}$  

Conclude: $\text{rank }\left(\sum_{k=1}^{n-1}\mu_k \cdot \big(E_{\lambda_{k+1}}-E_{\lambda_k}\big)\right)\lt \infty$ by sub-additivity of rank, thus  $\big\Vert T-\sum_{k=1}^{n-1}\mu_k \cdot \big(E_{\lambda_{k+1}}-E_{\lambda_k}\big)\big\Vert\leq \epsilon$   
tells us $T$ is approximated by a finite rank operator for arbitrarily small $\epsilon$, hence $T$ is compact [Theorem 4.10 in MacCluer]  


**Corollary:**  

If $S,T \in B(\mathscr H)$ and 
$S^*S\preceq T^*T$ where $T$ is compact, then $S$ must be compact as well.  

Consider arbitrary orthonormal sequence $\big\{e_k\big\}$  

$\big\Vert Se_k\big\Vert^2 =\big\langle S^*Se_k, e_k\big\rangle\leq \big\langle T^*Te_k, e_k\big\rangle =\big\Vert Te_k\big\Vert^2\to 0$  
where the upper bound goes to zero by ex 4.19 and by the above converse this means $S$ must be compact.  
[This is ex 9.3.14 in Kreyszig, albeit done out of order]  

note: it seems a more elementary answer is typically given / asked for.  In this case  consider a bounded sequence $x_k$ and by compactness of $T$ there is a convergent subsequence $T x_{k_j}$ so  

$\big\Vert S \big(x_{k_j}-x_{k_m}\big)\big\Vert^2 =\big\langle S^*S\big(x_{k_j}-x_{k_m}\big), \big(x_{k_j}-x_{k_m}\big)\big\rangle \leq \big\langle T^*T\big(x_{k_j}-x_{k_m}\big), \big(x_{k_j}-x_{k_m}\big)\big\rangle =\big\Vert T\big(x_{k_j}-x_{k_m}\big)\big\Vert^2\to 0$

**4.28**  
Are the Hilbert-Schmidt operators a closed subspace of $B(\mathscr H)$?  
The answer is no. The subspace of Hilbert-Schmidt operators are a *proper* subspace of compact operators, but if the subspace of Hilbert-Schmidt operators was closed then it would have to contain all compact operators hence couldn't be a proper subspace.  

In particular, recall that every compact operator on a Hilbert space is the limit of a sequence of finite-rank operators.  Every finite rank operator is necessarily Hilbert-Schmidt, so if the subspace of Hilbert-Schmidt operators was closed then the limit of that sequence, i.e. an arbitrary compact operator, would be Hilbert-Schmidt.  


As an explicit (counter-)example, consider diagonal operators  
$A^{(k)}$ where $A^{(k)}e_n = \sqrt{\frac{1}{n}}\cdot e_n$ for $1\leq n\leq k$ and $A^{(k)}e_n=\mathbf 0$ otherwise. Let $Ae_n = \sqrt{\frac{1}{n}}\cdot e_n$ for all $n\in \mathbb N$.   

$\sqrt{\frac{1}{n+1}}=\big\Vert A^{(k)}-A\big\Vert\to 0$   

so $A$, being the limit of finite rank operators, is compact but  
$\sum_{n=1}^\infty \big\Vert Ae_n\big\Vert^2=\sum_{n=1}^\infty \frac{1}{n}\not\lt \infty$  
[recall from Prop 4.14 (alternatively ref problem 8.5 in N. Young) that this is invariant to choice of Orthonormal Basis]  

Conclude: $A$ is not Hilbert-Schmidt and the subspace of Hilbert-Schmidt operators is not closed under limits, i.e. is not closed.  


**extension:**  
Working in a separable Hilbert space, if we change the norm from the operator norm to the Hilbert-Schmidt norm then the space of Hilbert-Schmidt operators is *complete* (hence closed under said limits, and a Hilbert space in its own write with inner product $\big\langle A,B\big\rangle_{HS} = \sum_{k=1}^\infty \langle Ae_k, Be_k\rangle$, etc.)  

Let $B(H)$ be the vector space of operators with the usual operator norm.  And let $V$ be the vector space of operators with Hilbert Schdmidt norm.  Suppose $V\ni\big\{A_k\big\}$ is Cauchy.  

Now for $\epsilon \gt 0$ we have for some unit vector $v$  
$\big\Vert A_k-A_j\big\Vert-\epsilon \leq \big\Vert (A_k-A_j)v\big\Vert\leq \big\Vert A_k-A_j\big\Vert_{HS}$   
[extend to orthonormal basis with $v$ being the 1st basis vector to get the upper bound] so $\big\{A_k\big\}$ is Cauchy with respect to the operator norm.  I.e. this shows that the Hilbert-Schmidt norm is stronger than the operator norm and necessarily $A_k\to A$ in $B(H)$ for some $A$ with respect to the operator norm, since $B(H)$ is complete, which implies $A_k\to_s A$ (strongly in $B(H)$)[recall ex 2.23].  With $\big(\mathbb N,\mu\big)$, the counting measure, and$f_k: \mathbb N\rightarrow \mathbb R$, given by $f_k\big(n\big)= \big\Vert A_ke_n\big\Vert^2$ and $\lim_{k\to \infty }f_k = f$ [point-wise, courtesy of strong convergence and continuity of norms and squaring] given by $f\big(n\big)= \big\Vert Ae_n\big\Vert^2$, where of course $\big\{e_n\big\}_{n=1}^\infty$ is some arbitrary orthonormal basis, then  

$$\sum_{n=1}^\infty \big\Vert A e_n\big\Vert^2 = \int f d\mu\leq \liminf_k \int f_k d\mu =\liminf_k \big\Vert A_k\big\Vert_{HS}\lt\infty \quad\text{(Fatou's Lemma)}\implies A\text{ is Hilbert-Schmidt}$$  
where $\Big\{\big\Vert A_k\big\Vert_{HS}\Big\}$ is Cauchy (reverse triangle inequality) hence bounded.  

Finally, select some orthonormal basis $\big\{e_k\big\}_{k=1}^\infty$ for $V$ and proceed by truncation.  I.e. by virtue of our sequence being Cauchy we have, for any $\epsilon \gt 0$ and $m, j\geq J^{(\epsilon)}$ and arbitrary $r \in \mathbb N$    
$\sum_{k=1}^r \big\Vert A_me_k - A_je_k\big\Vert^2 \leq \sum_{k=1}^\infty \big\Vert (A_m - A_j)e_k\big\Vert^2=\big\Vert A_m-A_j\big\Vert_{HS}^2 \leq \epsilon$  
$\implies \sum_{k=1}^r \big\Vert A e_k - A_je_k\big\Vert^2=\lim_{m\to \infty}\sum_{k=1}^r \big\Vert A_m e_k - A_je_k\big\Vert^2 \leq  \epsilon\quad\text{(take limit of m, using strong convergence on }B(H)\text{ )}$  
$\implies \big\Vert A-A_j\big\Vert_{HS}^2= \sum_{k=1}^\infty \big\Vert (A  - A_j)e_k\big\Vert^2 = \lim_{r\to\infty}\sum_{k=1}^r \big\Vert A e_k - A_je_k\big\Vert^2\leq  \epsilon\quad\text{(take limit of r)}$  
$\implies \lim_{j\to\infty}\big\Vert A-A_j\big\Vert_{HS}= 0$ as required.

Conclude: $\dim \ker (T-\lambda I)^k = \dim \ker (T-\lambda I)^{k+1}\implies \ker (T-\lambda I)^k=\ker (T-\lambda I)^{k+1}= \ker (T-\lambda I)^{k+2}=\ker (T-\lambda I)^{k+3}=\dotso$  

This argument for (ii.) relied on writing the vector space as a direct sum of $V$ plus something else-- it used $V^\perp$ though $V\oplus W$ for any closed $W$ would have sufficed.  The direct sum generalizes to Banach spaces (or normed spaces in general) **so long as** $\dim V\lt \infty$ since in general writing a direct sum of closed subspaces is not possible and raises technical issues, ref https://math.stackexchange.com/questions/108284/example-of-a-closed-subspace-of-a-banach-space-which-is-not-complemented , however we *can* do this in our case since $\dim V=d\lt \infty$.  One approach is to write a basis for $V\subseteq X$ with unit vectors $\big\{v_1, v_2, \dotso, v_d\big\}$ and via Hahn-Banach get bounded $f_j$ such that $f_j(v_k) =\delta_{j,k}$ since e.g. $\text{span}\big\{v_2, \dotso, v_d\big\}$ is closed, not containing $v_1$ hence $v_1$ has positive distance to it [use ex 3.8 / Corollary 3.6].  Then define $P:X\rightarrow X$ with $P(x)=\sum_{k=1}^d f_k(x)\cdot v_k$ and via triangle inequality $P$ is bounded plus algebraically check $P^2=P$, i.e. $P$ is a projection so $\text{span}\big\{v_1, v_2, \dotso, v_d\big\}=\text{image }P=\ker \big(I-P\big)$ is closed and so is $\ker P$ and $X = \text{image }P \oplus \ker P$ [ref Pryce 10S(iii) for the direct sum].  

**(iii.)** Schur Triangularization details  
While we are working in a more abstract setting, the argument runs essentially as usual.  At each step we have a basis ('hyper-vector' in the parlance of Artin)  
$\mathbf B = \bigg[\begin{array}{c|c|c|c} \mathbf B_1 & u_1 & \cdots &u_n\end{array}\bigg]$  
Where $\mathbf B_1$ is the pre-existing orthnormal basis and the $u_k$ any selection that extends that to an orthonormal [with respect to our Hilbert space's inner product]  basis for the new larger finite dimensional vector space we are considering  

*Technical nit:*  To sidestep awkwardness during step 1, after finding the unit length eigenvector, we may interpret this as being $\mathbf B_1 = \bigg[\begin{array}{c} v_1\end{array}\bigg]$.  

And by the work already done $T\mathbf B_1 = \mathbf B_1 R_1$ where $R_1$ is an upper triangular matrix.  
$\implies T\mathbf B = \mathbf B \left[\begin{matrix}R_1 & *\\\mathbf 0 & M\end{matrix}\right]= \left(\mathbf B \left[\begin{matrix}I & \mathbf 0 \\\mathbf 0 & Q\end{matrix}\right]\right)\left(\left[\begin{matrix}I & \mathbf 0 \\\mathbf 0 & Q\end{matrix}\right]^*\left[\begin{matrix}R_1 & *\\\mathbf 0 & M\end{matrix}\right]\left[\begin{matrix}I & \mathbf 0 \\\mathbf 0 & Q\end{matrix}\right]\right)\left[\begin{matrix}I & \mathbf 0 \\\mathbf 0 & Q\end{matrix}\right]^*=\left(\mathbf B \left[\begin{matrix}I & \mathbf 0 \\\mathbf 0 & Q\end{matrix}\right]\right)\left[\begin{matrix}R_1 & *\\\mathbf 0 & R_2\end{matrix}\right]\left[\begin{matrix}I & \mathbf 0 \\\mathbf 0 & Q\end{matrix}\right]^*$  
where $Q^*M Q=R_2$ for unitary $Q$ and upper triangular $R_2$ by the usual Schur Triangularization induction hypothesis.  


$\implies T\mathbf B' = T\mathbf B\left[\begin{matrix}I & \mathbf 0 \\\mathbf 0 & Q\end{matrix}\right] =\mathbf B' \left[\begin{matrix}R_1 & *\\\mathbf 0 & R_2\end{matrix}\right]$  

where $\mathbf B'=\mathbf B\left[\begin{matrix}I & \mathbf 0 \\\mathbf 0 & Q\end{matrix}\right]$ is still an orthonormal basis, recalling this result from Artin chapter 7.  E.g.  
$\left \langle \bigg[\begin{array}{c|c|c}  u_1 & \cdots &u_n\end{array}\bigg]\mathbf q_j, \bigg[\begin{array}{c|c|c}  u_1 & \cdots &u_n\end{array}\bigg]\mathbf q_r\right\rangle$  
$=\big\langle \sum_{k=1}^n q_{k,j}\cdot u_k, \sum_{m=1}^n q_{m,r}\cdot u_m\big\rangle$  
$=\sum_{k=1}^n q_{k,j}\cdot\sum_{m=1}^n \overline q_{m,r}\cdot\big\langle  u_k,  u_m\big\rangle$  
$=\sum_{k=1}^n q_{k,j}\overline q_{k,r}$  
$=\mathbf q_r^*\mathbf q_k$  
$=\delta_{k,r}$   
and those vectors live in the orthogonal complement of the span of $\mathbf B_1$ hence they are still orthogonal to the pre-existing orthonormal set.  At the end we re-label $\bigg[\begin{array}{c|c|c}  u_1 & \cdots &u_n\end{array}\bigg]\mathbf q_j$ to be the appropriate $v_i$.

**(iv.)** Generalizing the proof to Banach space $X$    
We want an ascending filtration of $T$ invariant subspaces where the difference in dimension $=1$ in each case, so re-write 

$V:=\bigcup_{k=1}^\infty \ker (T-\lambda I)^k=\bigcup_{k=1}^\infty W_k$  
where again by *contradiction* we assume $\dim V=\infty$.  However in this case we now insist that each $\dim W_k=k$.  [The one step increments of $W_k$'s dimension is not strictly needed, but at the cost of longer setup it makes that actual argument much easier to see.]  

We do this by triangularizing $T$'s restriction to $\ker (T-\lambda I)$ so $\mathbf B_1$ is a basis such that $T\mathbf B_1= \mathbf B_1 R$ for upper triangular $R$ [Artin chp. 4]. $\mathbf B' := \bigg[\begin{array}{c} \mathbf B_1 \end{array}\bigg]$ and $W_k$ for $1\leq k \leq \dim B_1$ is defined to be the span of first [left most] $k$ vectors in $\mathbf B'$. Then for the inductive step, consider $\ker (T-\lambda I)^m$  and  

$\mathbf B = \bigg[\begin{array}{c|c} \mathbf B_1 & *\end{array}\bigg]$  
where $*$ denotes arbitrary basis vectors chosen for $\mathbf B$.  Then 
$T\mathbf B =\mathbf B\left[\begin{matrix}R_1 & *\\\mathbf 0 & M\end{matrix}\right] $ and $M$ is similar to an upper triangular matrix $R_2$, so $\left[\begin{matrix}I & \mathbf 0 \\\mathbf 0 & S\end{matrix}\right]^{-1}\left[\begin{matrix}R_1 & *\\\mathbf 0 & M\end{matrix}\right] \left[\begin{matrix}I & \mathbf 0 \\\mathbf 0 & S\end{matrix}\right]  =\left[\begin{matrix}R_1 & *\\\mathbf 0 & R_2\end{matrix}\right]$  and  
$T\mathbf B' =\mathbf B'\left[\begin{matrix}R_1 & *\\\mathbf 0 & R_2\end{matrix}\right]=\bigg[\begin{array}{c|c|c|c|c} \mathbf B_1 & \mathbf B_2 \end{array}\bigg]\left[\begin{matrix}R_1 
& *\\\mathbf 0 & R_2\end{matrix}\right]$  
the same as with Schur Triangularization, except now we discard notions of orthogonality.  We do however insist each basis vector has length one. And as above, $W_k$ for $1\leq k \leq \dim B_1$ is defined to be the span of first [left most] $k$ vectors in $\mathbf B'$ (and notice this has an overlapping subproblem structure-- i.e. the first $k$ vectors for $\mathbf B'$ for $1\leq k\leq \dim B_1$ are those same $k$ vectors of $\mathbf B$ i.e. of $\mathbf B_1$).     

*Now we have inductively defined the desired* $W_k$, each of which satisfies $TW_k\subseteq W_k$.  We let $v_1$ be the (unit length) basis vector of $W_1$ which serves as a Base case.  


Inductive case, for $m\geq 2$  
Applying Riesz' Lemma [ref my corollary to ex 3.8 in 'MacCluer_chp3.ipynb'], there is some $v_k \in W_k$ such that $\big\Vert v_k\big\Vert=1$  and $d\big(v_k, W_{k-1}\big)\geq \frac{1}{2}$.   
[Given that $W_k$ is itself finite dimensional we could insist the distance is $=1$ though this is not needed.]     

$\mathbf B_k := \bigg[\begin{array}{c|c|c|c} v_1 & \dotso &v_{k-1} & v_k\end{array}\bigg]$  
$\implies T\mathbf B_k = \mathbf B_k\left[\begin{matrix}R & *\\\mathbf 0^T & \lambda \end{matrix}\right]$  
where $R$ is upper triangular by induction hypothesis and the lower right corner is necessarily $=\lambda$ since $(T-\lambda I)$ is nilpotent when restricted to $W_k$. And for $1\leq j\leq k-1$    
$\implies d\big(Tv_k,\big\{Tv_j\big\}\big)\geq  d\big(Tv_k, W_{k-1}\big)=\big\Vert Tv_k\big\Vert_{X\diagup  W_{k-1}}=\big\Vert \lambda \cdot v_k\big\Vert_{X\diagup  W_{k-1}}=\vert \lambda \vert \cdot \big\Vert  v_k\big\Vert_{X\diagup  W_{k-1}}\geq \vert \lambda \vert\cdot \frac{1}{2}$   
since $\big\{Tv_j\big\}\subseteq W_{k-1}$ and $Tv_k = \lambda \cdot v_k + w$ for some $w\in W_{k-1}$  

But we have constructed a sequence of unit length (hence bounded) $v_k\in X$ such that $\big\{Tv_k\big\}$ has no Cauchy subsequence (hence no convergent subsequence) since if we pick two arbitrary vectors $\big\{Tv_m, Tv_k\big\}$ with $k\gt m$ then we have  
$\big \Vert T v_k -Tv_m\big\Vert = d\big(Tv_k,\big\{Tv_m\big\}\big)\geq  d\big(Tv_k, W_{m}\big)\geq d\big(Tv_k, W_{k-1}\big)\geq\vert \lambda \vert\cdot \frac{1}{2}$   
since $\big\{Tv_k\big\}\subseteq W_m\subseteq W_{k-1}$ and this contradicts the compactness of $T$

**Corollary:**  
If $T \in K(\mathscr H)$ and $A\in B(\mathscr H)$  then $AT$ and $TA$ have the same non-zero eigenvalues with the same 'algebraic multiplicities'.  Recall of course since compact operators form an ideal so we know that $AT$ and $TA$ are both compact.    

Let $\lambda \in \mathbb C\setminus\big\{0\big\}$ and consider  
$W_m:=\ker \big(\lambda I-TA\big)^m$ then $A$ is injective on $W_m$ since $\big(\lambda I-TA\big)^m  =\lambda^m I+ p\big(TA\big)$ where $p$ is a polynomial with zero constant term so for $w\in W_m\cap \ker A$ then $\mathbf 0 = \big(\lambda I-TA\big)^m w=\lambda^m w + p\big(TA\big)w=\lambda^m w\implies w=\mathbf 0$.  
$\implies \ker \Big(A\big(\lambda I-TA\big)^m\Big)= \ker \Big(\big(\lambda I-AT\big)^m A\Big)$ 
so for non-zero $v \in \ker W_m$ we have  
$\mathbf 0= A\big(\lambda I-TA\big)^m v = \big(\lambda I-AT\big)^m \big(Av\big)$  
where $\big(Av\big)\neq \mathbf 0$ since $A$ is injective on $\ker \big(\lambda I-TA\big)^m$, i.e. $A$ injectively maps $\ker \big(\lambda I-TA\big)^m$ into $\ker \big(\lambda I-AT\big)^m$  
$\implies \dim\ker \big(\lambda I-TA\big)^m\leq \dim\ker \big(\lambda I-AT\big)^m\quad\text{(one cannot injectively map from a finite dim space into a lower dimensional space)} $  

And by mirroring the argument  
$\implies \dim\ker \big(\lambda I-AT\big)^m\leq \dim\ker \big(\lambda I-TA\big)^m $  

$\implies \dim\ker \big(\lambda I-AT\big)^m= \dim\ker \big(\lambda I-TA\big)^m $  

*remark:*  
We don't actually need to require that $T$ or $A$ is compact.  All we need is to know that $TA$ and $AT$ are compact (which can happen in special cases with non-compact $A,T\in B(\mathscr H)$ though typically we'd want to know before hand that one of $T,A$ is compact to ensure the products are compact).  


**Bonus problem**  
[from Halmos's *Hilbert Space Problem Book*]  

Suppose we have *normal* operator $A$ and $A^n$ is compact for some positive integer $n$.  Show that $A$ is compact. 

*Argument 1:*   
If $A^n$ is compact, then [by ex 4.6] we know $A^nA^j=A^{2^k}$ is compact for say for the smallest $k\in\mathbb N$ such that $2^k\geq n$. Thus [by ex 4.6] we know $(A^*A)^{2^k}=A^{2^k}(A^*)^{2^k}$ is compact, where the equality holds by normality of $A$.  

From the top of this notebook, we know $T \text{ is compact }\impliedby T^*T\text{ is compact }$ thus  
$(A^*A)^{2^{k}}$ is compact $\implies (A^*A)^{2^{k-1}}$ is compact which in turn implies $(A^*A)^{2^{k-2}}$ is compact and so on until [formally via induction] we deduce $A^*A$ is compact which implies $A$ is compact.  



*Argument 2:*   
Note if $A^n$ is compact then $A^rA^n$ is compact for all $r\in \mathbb N$ [compact ideals] so via well chosen $r$ is suffices to consider the case where $n=2^k$ for some $k\in \mathbb N$.   

Using Polar Decomposition, $A=UP=PU$ [Ref Brezis Problem 44(6)] where $U$ is unitary, $UP=PU$ and $P=\sqrt{A^*A}$  

Base Case: $n=2$ (i.e. $k=1$)  
suppose $A^2= U^2 P^2=U^2 A^*A$ is compact $\implies U^{-2}A^2=A^*A=P^2$ is compact 
$\implies P=\sqrt{A^*A}$ is compact  
['Diagonalize' $P^2$ and take square-root of eigenvalues to get diagonalized $P$ which is a limit of finite rank operators; recall uniqueness of positive square-root.]   
$\implies A=UP$ is compact since compact operators form an ideal  

Inductive Case (induct on $k$):  
$n=2^k$ for some $k\geq 2$  
$B:=A^{\frac{n}{2}}=A^{2^{k-1}}$  
Then $B$ is normal [e.g. it has polar decomposition $B=U^{\frac{n}{2}}P^{\frac{n}{2}}=P^{\frac{n}{2}}U^{\frac{n}{2}}$ where $U^\frac{n}{2}$ is unitary hence $B$ is normal]

$B^2$ is compact and normal hence $B=A^{2^{k-1}}$ is compact by the Base Case argument and $A^{2^{k-1}}$ being compact $\implies A$ is compact by induction hypothesis.
