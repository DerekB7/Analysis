**10.9.2**  
In the space $\mathscr l^1$ with the usual norm define for $x= (x_i)_{i=1}^\infty$ and $y= (y_i)_{i=1}^\infty$ the (component-wise) product $x\cdot y= (x_i\cdot y_i)_{i=1}^\infty$.  Prove that $\mathscr l^1$ is a Banach algebra without the identity.  

$\mathscr l^1$ is a Banach space, the product is associative $(x\cdot y)\cdot z= \begin{bmatrix}x_1 \\ x_2  \\ x_3\\\vdots \end{bmatrix}\circ \begin{bmatrix}y_1\cdot z_1 \\ y_2 \cdot z_2 \\ y_3\cdot z_3 \\\vdots \end{bmatrix}=\begin{bmatrix}x_1\cdot y_1\cdot z_1 \\ x_2 \cdot y_2\cdot z_2 \\ x_3\cdot y_3\cdot z_3 \\\vdots \end{bmatrix} =\begin{bmatrix}x_1\cdot y_1 \\ x_2 \cdot y_2 \\ x_3\cdot y_3 \\\vdots \end{bmatrix}\circ\begin{bmatrix}z_1 \\ z_2  \\ z_3\\\vdots \end{bmatrix}=x\cdot(y\cdot z) $
and similarly   
$(\alpha x)\cdot y = \begin{bmatrix}\alpha \cdot x_1 \\ \alpha \cdot x_2  \\\alpha \cdot x_3\\\vdots \end{bmatrix}\circ \begin{bmatrix}y_1 \\ y_2 \\ y_3 \\\vdots \end{bmatrix}=\begin{bmatrix}\alpha \cdot x_1\cdot y_1 \\\alpha \cdot x_2 \cdot y_2 \\ \alpha \cdot x_3\cdot y_3 \\\vdots \end{bmatrix} =\alpha\cdot \begin{bmatrix}x_1\cdot y_1 \\ x_2 \cdot y_2 \\ x_3\cdot y_3 \\\vdots \end{bmatrix}= \alpha\cdot (x\cdot y)=x\cdot(\alpha \cdot y)$  
where we could directly  verify the last equality or use the fact that this point-wise (Hadamard) product is commutative.  

And we may verify $x\cdot y\in \mathscr l^1$, and that the norm is sub-multiplicative, since $y\in \mathscr l^1\implies y\in \mathscr l^\infty$ hence $\big\Vert x\cdot y\big\Vert_1 \leq \big\Vert x\big\Vert_1\cdot \big\Vert y\big\Vert_\infty$  [closely related to the dual space argument]   
$\implies \big\Vert x\cdot y\big\Vert_1 \leq \big\Vert x\big\Vert_1\cdot \big\Vert y\big\Vert_\infty\leq \big\Vert x\big\Vert_1\cdot \big\Vert y\big\Vert_1$

I directly show the continuity of the product, below, though as noted by MacCluer (p. 107) a sub-multiplicative norm will force the product to be continuous.  Consider $\Vert x^{(n)}- x\Vert \to 0$ and $\Vert y^{(n)}- y\Vert \to 0$ and we need to show $\Vert x^{(n)}\cdot y^{(n)}- x\cdot y\Vert \to 0$. 

$\big\Vert x^{(n)}\cdot y^{(n)}- x\cdot y\big\Vert =\big\Vert x^{(n)}\cdot y^{(n)}-x^{(n)}\cdot y+ x^{(n)}\cdot y- x\cdot y\big\Vert$  
$\leq \big\Vert x^{(n)}\cdot (y^{(n)}- y)\big\Vert+\big\Vert (x^{(n)}-x)\cdot y\big\Vert$  
$\leq \big\Vert x^{(n)}\big\Vert_1 \cdot \big\Vert y^{(n)}- y)\big\Vert_\infty+\big\Vert x^{(n)}-x\big\Vert_1\cdot \big\Vert y\big\Vert_\infty$  
$\leq \big\Vert x^{(n)}\big\Vert_1 \cdot \big\Vert y^{(n)}- y)\big\Vert_1+\big\Vert x^{(n)}-x\big\Vert_1\cdot \big\Vert y\big\Vert_\infty$  
and taking limits gives an upper bound of zero since by continuity of the norm (and products in $\mathbb R$) we get $\big\Vert x^{(n)}\big\Vert_1\to \big\Vert x\big\Vert_1$ and $ \big\Vert y^{(n)}- y)\big\Vert_1\to 0$ and $\big\Vert x^{(n)}-x\big\Vert_1\to 0$  

However this must be a Banach Algebra without the identity since if there was an identity $e$ it would satisfy  
$e\cdot \mathbf e_k =\mathbf e_k\implies $ its kth component is $1$ hence $\sum_{k=1}^\infty 1\lt \infty$ which is absurd.

Let $e, I$ be the identity elements.  Then $(I-e)x=0$ for all $x\in \mathscr A$ set $x:=e$ and  
$I-e = Ie-ee=(I-e)e=0\implies I=e$  


**(b.)** Prove that if an element of an algebra is invertible then its inverse is unique  
Let $c,d $ be inverses of $a\in \mathscr A$.  And since $a$ is invertible then each $x\in \mathscr A$ may be written as $x =ay$ for some $y\in \mathscr A$ since $x \in a\mathscr A$.

$0=(c-d)e=c-d\implies c=d$

Let $\mathscr A$ be a Banach algebra.  Prove the following  
(a.) If $x\in\mathscr A$, $\Vert x\Vert \lt 1$ then  
$\big\Vert (e-x)^{-1}-e -x\big\Vert \leq \frac{\Vert x\Vert^2}{1-\Vert x\Vert}$  
This is a geometric / Neumann series argument, i.e.  
$\big\Vert (e-x)^{-1}-e -x\big\Vert=\big\Vert \big(e + x + x^2 + x^3+ x^4+ \dots\big)-e -x\big\Vert=\big\Vert x^2\big(e+x+x^2\big)\big\Vert\leq \big\Vert x^2\big \Vert \cdot \big(\big\Vert e\big\Vert +\big\Vert x\big\Vert +\big\Vert x\big\Vert^2\dots\big)=\frac{\Vert x\Vert^2}{1-\Vert x\Vert}$  

(b.) If $x\in\mathscr A$ is invertible and $y\in \mathscr A$ and $\Vert x-y\Vert \lt \frac{1}{\Vert x^{-1}\Vert}$ then y is invertible and  
$\big\Vert x^{-1}- y^{-1}\big\Vert \leq \frac{\Vert x^{-1}\Vert^2\cdot \Vert x-y\Vert}{1-\Vert  x^{-1}\Vert \cdot \Vert x-y\Vert}$  

**ref Chapter 5, in chapter exercise 1 in Retherford.ipynb**  
That exercise was nominally done in the setting of $B(H)$, bounded operators on a Hilbert space, which is an example of a Banach Algebra, but the operations all carry through verbatim to the more general setting of Banach Algebras.  

*remark:*  
part $b$ tells us that the set of invertible elements is open (hence the spectrum is closed), since any invertible element $x$ has a positive norm, and any $y\in B\big(x, \Vert x^{-1}\Vert^{-1}\big)$ is invertible. It also tells us that if $\Vert x_k- x\Vert\to 0$ then not only are all $x_k$ invertible for $k$ large enough but      
$\big\Vert x^{-1}- x_k^{-1}\big\Vert \leq \frac{\Vert x^{-1}\Vert^2\cdot \Vert x-x_k\Vert}{1-\Vert  x^{-1}\Vert \cdot \Vert x-x_k\Vert}\leq \frac{\Vert x^{-1}\Vert^2\cdot \Vert x-x_k\Vert}{1-\Vert  x^{-1}\Vert}\to 0$  
where the 2nd inequality holds for $k\geq K$.  This means that the inverse is a continuous operation.  

**10.9.6**  
Let $O$ be the set of invertible elements of a Banach algebra $\mathscr A$ and let $x$ be a boundary point of $O$, i.e. $x\in \overline B\text{ \ }\stackrel{o}{B}$  

(a.) Prove that if $O\ni x_n$ for all $n$ and $x_n\to x$ then $\Vert x_n^{-1}\Vert \to \infty$  
per the remark of the prior exercise, the set of invertible elements is open hence if $x\in \overline B\text{ \ }\stackrel{o}{B}$, then $x$ is not invertible. Now suppose for contradiction that $\Vert x_n^{-1}\Vert \not\to \infty$ then that for some $M\gt 0$ we have a subsequence $\Vert x_{n_k}^{-1}\Vert\lt M$ (infinitely often) but $\Vert x_{n_k}- x\Vert \to 0$ hence there is some  
$\Vert x_{n_K}- x\Vert\leq \frac{1}{M}\lt \frac{1}{\Vert x_{n_k}^{-1}\Vert}$ and $x$ is invertible after all, a contradiction.  

(b.) Prove that there exists a sequence $y_n$ such that $\Vert y_n\Vert =1$, $xy_n\to 0$ and $y_nx\to 0$.  
Set $y_n:=\frac{x_n^{-1}}{\Vert x_n\Vert}$ and for any $\epsilon \gt 0$ for all $n$ large enough we have    

$\big\Vert y_n x\big\Vert  -\big\Vert y_nx_n\big\Vert\leq \big\Vert y_n x -y_nx_n\big\Vert \leq\big\Vert y_n\big\Vert \cdot \big\Vert x-x_n\big\Vert = \big\Vert x-x_n\big\Vert\lt \epsilon $  
$\implies \big\Vert y_n x\big\Vert \leq \epsilon+\big\Vert y_nx_n\big\Vert=\epsilon+\frac{1}{\Vert x_n\Vert}\big\Vert x_n^{-1}x_n\big\Vert=\epsilon+\frac{1}{\Vert x_n\Vert}\implies \limsup_n\big\Vert y_n x\big\Vert \leq \epsilon+\limsup_n\frac{1}{\Vert x_n\Vert}=\epsilon$  
$\implies \lim_{n\to\infty}\big\Vert y_n x\big\Vert=0$  
since $\epsilon\gt 0$ was arbitrary  

And re-running the argument this time for $xy_n$ instead of $y_n x$ yields  
$\big\Vert  xy_n\big\Vert  -\big\Vert x_ny_n\big\Vert\leq \big\Vert  xy_n -x_n y_n\big\Vert \leq  \big\Vert x-x_n\big\Vert\cdot\big\Vert y_n\big\Vert = \big\Vert x-x_n\big\Vert\lt \epsilon $  
$\implies \big\Vert xy_n \big\Vert \leq \epsilon+\frac{1}{\Vert x_n\Vert}\big\Vert x_n^{-1}x_n\big\Vert=\epsilon+\frac{1}{\Vert x_n\Vert}\implies \limsup_n\big\Vert xy_n \big\Vert \leq \epsilon+\limsup_n\frac{1}{\Vert x_n\Vert}=\epsilon$  
$\implies \lim_{n\to\infty}\big\Vert  xy_n\big\Vert=0$  


**10.9.7**  
Prove that if for a [unital, non-trivial] Banach Algebra $\mathscr A$ there exists $M\lt \infty$ such that for any $x,y\in \mathscr A$ we have $\Vert x\Vert\cdot\Vert y\Vert \leq M\cdot\Vert xy\Vert$ then the algebra $\mathscr A$ is isometrically isomorphic to the field $\mathbb C$  

This is a slight variation on Gelfand-Mazur.  I give a MacCluer-based solution, though the ideas are related to exercise 10.9.6.  

Suppose $x \not \propto e$.  Then $x':=x-\lambda e$ for $\lambda \in \mathbb C$ is bounded below, i.e. we cannot have a sequence $\big\{y_k\big\}$ where $\big\Vert y_k\big\Vert=1$ for $k\in \mathbb N$ such that   
$\frac{1}{M}\cdot\big\Vert x'\big\Vert=\frac{1}{M}\cdot\big\Vert x'\big\Vert\cdot\big\Vert y_k\big\Vert \leq \big\Vert x'y_k\big\Vert\to 0\quad \text{(absurd)}$  

Thus $x$ has no approximate point spectrum.  Now recalling the fact that $\emptyset\neq \sigma(x)$ is compact, so it includes its boundary, we have  
$\emptyset\neq\partial\sigma(x)\subseteq \sigma(x)$  
hence there is $\lambda \in \partial\sigma(x)$ and that $\lambda$ necessarily is an approximate eigenvalue [MacCluer ex 5.5(d)], a contradiction.  

Thus for each $x\in \mathscr A$ we have $x \propto e$, i.e. $x=\lambda\cdot e$, where since $\mathscr A$ is closed under re-scaling, necessarily there is an $x$ for each $\lambda  \in \mathbb C$ 


**10.9.8**  
[mentioned on page 177 as that treats the commutative case and this is for the general non-commutative case though it seems that more or less the same argument as there applies here]  

Let $\mathscr A$ be a Banach algebra and let $\varphi :\mathscr A\rightarrow \mathbb C$ be a multiplicative functional, i.e. $\varphi$ is non-trivial and linear, such that $\varphi(x\cdot y)=\varphi(x)\cdot \varphi(y)$ for any $x,y\in \mathscr A$ [in effect: a functional that is also a ring homomorphism].  Prove the following:   

(a.) $\varphi(e)=1$ and $\varphi(x)\neq 0$ for any invertible $x$.  
$\varphi(e)= \varphi(e\cdot e)=\varphi(e)\cdot \varphi(e)$  
$\implies \varphi(e)\in \big\{0,1\big\}$ (idempotence)  
and $\varphi(e)=0\implies \varphi(x)=\varphi(e\cdot x)=\varphi(e)\cdot \varphi(x)=0\cdot \varphi(x)=0$ for all $x \in \mathscr A$ hence $\varphi(e)=0\iff \varphi =0$ but by assumption $\varphi$ isn't the zero functional  

So for invertible $x$ we have  
$1=\varphi(e)=\varphi(x\cdot x^{-1})= \varphi(x)\cdot \varphi(x^{-1})\implies \varphi(x^{-1})=\varphi(x)^{-1}$  
$\implies\varphi(x)\neq 0$  


(b.) If $x\in \mathscr A$ and $\Vert x\Vert \lt 1$ then $\vert \varphi(x)\vert \lt 1$   
re-visiting the Neumann series  
$y_n= e + x + x^2 + \dots + x^{n-1}\implies (e-x)\cdot y_n=e -x^{n}$  
$\implies (e-x)y=(e-x)\cdot\lim_{n\to\infty } y_n=\lim_{n\to\infty }(e-x)\cdot y_n=e $  
where $\Vert x^n\Vert \leq \Vert x\Vert^n\to  0$ so $y=\lim_{n\to\infty} y_n= e + x + x^2 + \dots + x^{n-1}+\dots$  
where the series converges absolutely (and $\mathbb C$ is complete so the partial sums form a Cauchy sequence and hence converge)  

Thus $(e-x)^{-1}$ exists, and so does $(e-w\cdot x)^{-1}$ for any $w \in\overline B(0,1)$.  
But if $\vert \varphi(x)\vert\geq 1$ then there is a $w \in \overline B(0,1)$ thus that $\varphi(w\cdot x)= w\cdot \varphi(x)=1$ yielding   
$0=1 -1= \varphi(e)-\varphi(w \cdot x) = \varphi(e-w \cdot x)\neq 0$ which is absurd.  

Note: the technique of re-scaling by some $w$ in the closed unit disc is a common one in complex analysis type problems.  Ref the Gauss-Lucas based Bernstein Inequality proof e.g. in here https://math.stackexchange.com/questions/4688731/a-complex-analysis-proof-of-the-extremal-case-of-bernsteins-inequality/  

(c.) $\Vert \varphi\Vert =1$  
$\Vert \varphi\Vert \leq 1$  follows almost immediately since from (b.) if $\Vert x\Vert =1\implies \vert \varphi(x)\vert \leq 1$ since for any $\delta\in (0,1)$ we have $\delta\cdot \vert \varphi( x)\vert=\vert \varphi(\delta\cdot x)\vert \lt 1\implies \vert \varphi( x)\vert\leq \frac{1}{\delta}$ and take a limit as $\delta\to 1^-$    

and by (a.) $\varphi(e)=1$ where $\Vert e\Vert =1$ so $\Vert \varphi\Vert \leq 1$ is met with equality  




**10.9.9**  
Let $\mathscr A$ be a Banach algebra and let $x\in \mathscr A$, let $\mathbb C\supseteq\Omega$ be an open set and $\sigma(x)\subseteq \Omega$.  Prove that there exists $\delta\gt 0$ such that $\sigma(x+y)\subseteq \Omega$ for any $y\in \mathscr A$ such that $\Vert y\Vert \lt \delta$.  

$\Omega^C$ is closed but not bounded though we may truncate and consider $K:=\Omega^C\cap \overline B\big(0,1+\Vert x\Vert\big)$ which is the intersection of 2 closed sets, hence closed, and is a subset of $\overline B\big(0,1+\Vert x\Vert\big)$ hence bounded, i.e. we have compact $K\subseteq \mathbb C$. We have $(\lambda e-x)$ is invertible for $\lambda \in K$ so by ex 10.9.5(b) if $\zeta \in B\big(\lambda,\delta^{(\lambda)}\big)$ then $(\zeta e -x)$ is invertible [i.e. invertible elements live in an open set in a metric space  hence there is an open ball around each invertible element ('point') such that everything in that ball is invertible].  


Now $K\subseteq \bigcup_{\lambda\in K}\delta^{(\lambda)}$ is an open covering of a compact set hence it has a finite subcovering and a Lebesgue number $\delta'$.  Consider $\lambda \in \Omega^C\setminus K$ --which implies $\vert \lambda \vert \gt \Vert x\Vert +1$-- so if $\Vert y\Vert \lt 1$ we know $\big(\lambda e + (x+y)\big)$ is invertible since $\big\Vert x+y\big\Vert \lt \big\Vert x\big\Vert +1 \lt \vert \lambda\vert$ hence it is invertible by the Neumann series argument.  Set $\delta:=\min\big(\delta',1\big)$ and if $\Vert y\Vert \lt \delta$ then every point in $\Omega^C$ is regular for $(x+y)$ therefore $\sigma(x+y)\subseteq \Omega$.   

*remark:*  
This gives rough notions about spectral continuity in that the set of spectrum as a whole must lives in a neighborhood of the original spectrum after a perturbation.

**10.9.15**
#### I completely re-cut this question on the holomorphic functional calculus to factor most of the results as limits of rational functionals.    

Note:  
The book by Salamon and Buehler has a *very good* overview of the Holomorphic functional calculus, page 227, here https://people.math.ethz.ch/~salamon/PREPRINTS/funcana.pdf though the reference to why a suitable cylce always exists is broken; it mentions page 139 [and 140] of Ahlfors 3rd edition which seems to be erroneous and also a document written by Salamon that is only in German though the alleged page number is blank so that is erroneous too.  However, as noted below, Haslinger has a proof.  



Let $x$ be an element of a Banach algebra $\mathscr A$ and let $\mathbf F(x)$ be the set of functions $f$ which are analytic in a neighborhood of $\sigma(x)$ [the neighborhood may be non-connected and may depend on $f \in \mathbf F(x)$].  Let $f$ be an analytic function in a neighborhood $D$ of $\sigma(x)$ and let $U$ be an open set with its boundary $[\Gamma]$ being a finite union of [the image of] regular piecewise smooth curves such that $U\supseteq \sigma(x)$ and $U\cup \Gamma\subseteq D$.  Define the element $f(x)$ as follows  


$$f(x)=\frac{1}{2\pi i}\int_\Gamma f(\lambda)(\lambda e-x)^{-1}d\lambda$$  

Note:  ex 10.9.9 implies $f$ is continuous on $\mathscr A$, since if $\Vert x_k -x\Vert \to 0$ then do an open covering of $\sigma(x)$ such that the covering is all 'inside' $\Gamma$ and for $k$ large enough $\sigma(x_k)$ is 'inside' $\Gamma$ as well per ex 10.9.9 [note the below reference to Haslinger for some of the topological details not touched on in this answer that were skipped over in Eidelman, Milman & Tsolomitis as well], thus we can use the same spectral integral and  
$$\big\Vert f(x)-f(x_k)\big\Vert=\left\Vert \frac{1}{2\pi i}\int_\Gamma f(\lambda)\big((\lambda e-x)^{-1}-(\lambda e-x_k)^{-1}\big)d\lambda\right\Vert\to 0$$  
since $\big\Vert\big((\lambda e-x)^{-1}-(\lambda e-x_k)^{-1}\big)\big\Vert \to 0$ [per ex 10.9.5(b), using that $\Vert (\lambda e-x)^{-1}\Vert$ is a continuous function on a compact set, $[\Gamma]$ hence it has a max]  


**brief comment on power series**  
if $f(z) =\sum_{k=0}^\infty \alpha_k z^k$ [power series expansion at zero] or $f(z) =\sum_{k=0}^\infty \alpha_k (z-w)^k$ [power series expansion at w]  
and $\frac{1}{r_f}=\limsup_{n}\vert a_n\vert^\frac{1}{n}$ and $r(A)=\limsup_{n}\vert A\vert^\frac{1}{n}$ [from Theorem 10.3.5 is spectra radius], then if  
$F(z) =\sum_{k=0}^\infty \alpha_k  \Vert A^k\Vert\cdot z^k $ then  
$\frac{1}{r_F}=\limsup_{n}\big(\vert a_n\vert\cdot \vert \Vert A^n\Vert\vert\big)^\frac{1}{n}=\limsup_{n}\vert a_n\vert^\frac{1}{n}\Vert A^n\Vert^\frac{1}{n}\leq \limsup_{n}\vert a_n\vert^\frac{1}{n}\limsup_{n}\Vert A^n\Vert^\frac{1}{n}=\frac{r(A)}{r_f}$ 
so as long as $r(A)\lt r_f$ then $r_F=\frac{r_f}{r(A)}\gt 1$ and in particular $F(1)$ converges absolutely and its arbitrarily small tail (after ignoring a large enough number of initial terms) allows us to conclude implies $f_m(A) =\sum_{k=0}^m \alpha_k  A^k$ is a Cauchy sequence in $A$ so $f(A) =\sum_{k=0}^\infty \alpha_k  A^k$ converges.  That is the key reason why we want the power series to have a radius of convergence greater than $r(A)$-- in general strictly greater unless we want to get in the weeds of boundary behavior.

Cauchy Integral Inequality -- **ML-inequality implications**    
Using Runge, suppose we have $r_n\to f$ uniformly on some compact $K\in D$, where $\sigma(x), [\Gamma]\in K$, then $g_n(z):=f(z)-r_n(z)$ and for any $\epsilon \gt 0$,     

$$\big\Vert g_n(x)\big\Vert=\left \Vert \frac{1}{2\pi i}\int_\Gamma g_n(\lambda)(\lambda e-x)^{-1}d\lambda\right\Vert \leq L(\Gamma)\cdot \epsilon \cdot \max_{\lambda \in [\Gamma]}\left \Vert(\lambda e-x)^{-1}\right\Vert\quad\text{(arbitrarily small)} $$  

which we can partially unpack as  
$$\left \Vert \frac{1}{2\pi i}\int_\Gamma g_n(\lambda)(\lambda e-x)^{-1}d\lambda\right\Vert=  \left \Vert\frac{1}{2\pi i }\sum_{\gamma\in\Gamma}\int_\gamma g_n(\lambda)(\lambda e-x)^{-1}d\lambda\right\Vert\leq \sum_{\gamma\in\Gamma}\left \Vert \frac{1}{2\pi i }\int_\gamma g_n(\lambda)(\lambda e-x)^{-1}d\lambda\right\Vert$$

and to bound each term in $\sum_{\gamma\in\Gamma}\left \Vert \frac{1}{2\pi i }\int_\gamma g_n(\lambda)(\lambda e-x)^{-1}d\lambda\right\Vert$ we consider, since a cycle has finitely many ($m$) curves in it then     
$$\left \Vert \frac{1}{2\pi i}\int_\gamma g_n(\lambda)(\lambda e-x)^{-1}d\lambda\right\Vert \leq  \frac{1}{2\pi }\int_a^b \vert \gamma'(\lambda)\vert  \cdot\vert  g_n(\lambda)\vert\cdot \left \Vert(\lambda e-x)^{-1}\right\Vert d\lambda\leq L(\gamma)\cdot \frac{\epsilon}{m} \cdot \max_{\lambda \in [\Gamma]}\left \Vert(\lambda e-x)^{-1}\right\Vert\quad\text{(for }n\geq N^{(\gamma)}\text{)} $$  

I.e. to justify the ML-inequality here first unpack the cycle into the close curves $\gamma$ and apply triangle inequality.  Then on the final line we derive ML-inequality for an individual $\gamma$ as follows.  The 1st inequality is triangle inequality since we are essentially using the Riemann integral (so apply triangle inequality to Riemann sums / sequences of Riemann Partitions). And the 2nd inequality is ML-inequality.  By construction we have $\vert g_n(z)\vert \lt\epsilon$ for all $z\in K$. And note that the mapping $\big\{x,e\big\}\times \Gamma \rightarrow \mathbb R$  given by $(x,e,\lambda) \mapsto \left \Vert(\lambda e-x)^{-1}\right\Vert$  is the composition of scaling and addition  (continuous in the product topology) then inversion which is continuous per ex 10.9.5(b) then application of the norm function which is continuous and hence the mapping is a  composition of continuous functions so is itself continuous.  The domain is the product of compact sets [note: $\{x,e\}$ is a set in a metric space and it has the Bolzano-Weierstrass property any bounded sequence in it has a trivially convergent subsequence since at least one of $x$ or $e$ must be repeated infinitely often] hence is compact hence its image under this (continuous) mapping is compact and the $\max_{\lambda\in[\Gamma]}\left \Vert(\lambda e-x)^{-1}\right\Vert$ exists. 



To develop the integral-- i.e. show that it does what we want it to do, I find it best to start with algebraic objects-- polynomial functions and then their generalization and show that the integral representation agrees with them.  Beardon's book *Complex Analysis* takes this approach however it generalizes from polynomials to power series (Weierstrass approach) which works perfectly for scalar variable complex analysis; that approach has some radius of convergence issues in the Banach space setting so generalizing to rational functions which approximate analytic functions (Runge) is instead that approach I take here.  

**motivating example with square matrices**  
Remark: for concreteness, the motivation for this should be matrices $A\in M_n\big(\mathbb C\big)$ where $\sigma(A)$ is in the Domain of analytic $f$.  Start with the Runge notion of analytic functions being the uniform limit of rational functions (subject to certain details).  

With the linear map $T:M_n(\mathbb C)\rightarrow M_n(\mathbb C)$ given by $T(X)= SXS^{-1}$ (which is actually a ring homomorphism) and rational function $r$)    

$T\big(r(A)\big)= T\Big(p\big(A\big)q\big(A\big)^{-1}\Big)=p\big(T(A)\big)q\big(T(A)\big)^{-1}=r\big(T(A)\big)$  
and since $T$ is continuous, if [via Runge] a sequence of rational functions $r_n\to f$, analytic, on a compact set within the domain of $f$ (that contains the spectrum of $A$) then    

$T\Big(f\big(A\big)\Big)= T\Big(\lim_{n\to\infty}r_n\big(A\big)\Big)= \lim_{n\to\infty}T\Big(r_n\big(A\big)\Big)=\lim_{n\to\infty}r_n\Big(T\big(A\big)\Big)=f\Big(T\big(A\big)\Big)$  
so analytic $f$ should respect similarity transforms.  And we thus take $A$ to be WLOG block diagonal, as $\oplus_{j=1}^m\ker\big(\lambda_j I -A\big)^n$ [where $A$ has $m$ distinct eigenvalues, so in effect its generalized eigenspaces]  


$A=\begin{bmatrix}  A_1 &\mathbf 0 &\dots &\mathbf 0\\\mathbf 0& A_2 &\dots &\mathbf 0\\  \vdots  &  \vdots&\ddots&\vdots\\ \mathbf 0&\mathbf 0&\dots &A_m\end{bmatrix}$  

if invertible, then 
$A^{-1}=\begin{bmatrix}  A_1^{-1} &\mathbf 0 &\dots &\mathbf 0\\\mathbf 0& A_2^{-1} &\dots &\mathbf 0\\  \vdots  &  \vdots&\ddots&\vdots\\ \mathbf 0&\mathbf 0&\dots &A_m^{-1}\end{bmatrix}$ and for polynomial $p$ we have   $p\big(A\big)=\begin{bmatrix}  p\big( A_1\big) &\mathbf 0 &\dots &\mathbf 0\\\mathbf 0& p\big( A_2\big) &\dots &\mathbf 0\\  \vdots  &  \vdots&\ddots&\vdots\\ \mathbf 0&\mathbf 0&\dots &p\big(A_m\big)\end{bmatrix}$  


$\implies f\big(A\big)=\lim_{n\to\infty}r_n\big(A\big)=\lim_{n\to\infty}\begin{bmatrix} r_n\big( A_1\big) &\mathbf 0 &\dots &\mathbf 0\\\mathbf 0& r_n\big( A_2\big) &\dots &\mathbf 0\\  \vdots  &  \vdots&\ddots&\vdots\\ \mathbf 0&\mathbf 0&\dots &r_n\big(A_m\big)\end{bmatrix}=\begin{bmatrix} f\big( A_1\big) &\mathbf 0 &\dots &\mathbf 0\\\mathbf 0& f\big( A_2\big) &\dots &\mathbf 0\\  \vdots  &  \vdots&\ddots&\vdots\\ \mathbf 0&\mathbf 0&\dots &f\big(A_m\big)\end{bmatrix}$  

Thus analytic $f$ should be applied to each eigenspace (/block) restriction.  

Now $A_j$ has only eigenvalue $\lambda_j$ so do a local power series expansion at $\lambda_j$ so $f(z) =\sum_{k=0}^\infty c_k (z-\lambda_j)^k$ and apply $f\big(A_j\big)=\sum_{k=0}^\infty c_k (A_j-\lambda_j I)^k=\sum_{k=0}^{n-1} c_k (A_j-\lambda_j I)^k$ where the right hands side holds since $(A-\lambda_j I)$ is nilpotent.  

Now that we understand $f$ both in terms of local power series applications and as a limit of rational functions, we may consider it in terms of the Cauchy Integral Formula with a small closed curve $\gamma$ around each eigenvalue $\lambda_j$ and see it agrees with the power series approach for $f(A_j)$.And finally we can aggregate all of those closed curves $\gamma$ into a big cycle $\Gamma$ to recover the above formula. **review of partial fractions:**  
First, some review of rational functions and partial fractions is in order.  


Let $r(z)=\frac{p(z)}{s(z)}$  where $p$ and $s$ have no common roots.  The best case is when $r$ is *strictly proper*, i.e. when $\text{degree }p\lt \text{degree }s$.  If this is not the case then we can always consider $r(z)-q(z)$ for well chosen polynomial $q$ such that $(r-q)$ is strictly proper.  So the below assumes $r$ is strictly proper. WLOG take $s$ to be monic (i.e. 'move' the leading coefficient up to $p$ is needed).   

*Nice Case:*    
suppose all roots of $s$ are distinct / simple.  I.e. $s(z)= \prod_{k=1}^n (z-\lambda_k)$ with $n$ distinct roots $\big\{\lambda_1, \dotso,\lambda_n\big\}$ (equivalently $s$ has non-zero discriminant).  Then we may write  

$r(z) =\sum_{k=1}^n \frac{\alpha_k}{z-\lambda_k}$  
where we multiply each side by $(z-\lambda_k)$ and take a limit $z\to \lambda_k$ to deduce $\alpha_k$.  We can confirm that the RHS representation is in fact valid by multiplying each side by $s(z)$ (and eliminating removable singularities).  
$p(z)=s(z)\cdot r(z) =s(z)\sum_{k=1}^n \frac{\alpha_k}{z-\lambda_k}=\sum_{k=1}^n \alpha_k\prod_{j\neq k}(z-\lambda_j)$  

Both sides are polynomials and for the LHS we have $\text{degree }p\lt \text{degree }s=n$ and the RHS has degree $\leq n-1 \lt n$.  And they both agree for $n$ distinct points $\big\{\lambda_1, \dotso,\lambda_n\big\}$ hence they are the same polynomial (recall the Vandermonde writeup in the Linear Algebra folder for more info).  Then divide each side by $s(z)$ to recover $r(z) =\sum_{k=1}^n \frac{\alpha_k}{z-\lambda_k}$.  

*Ugly Case:*  
[rough overview]  
The problem is if we attempt the above argument and suitably clear denominators, we now have polynomials with degree at most $n-1$ on each side but only know they agree on $\leq n-1$ points which isn't enough to confirm they are the same polynomials so we need to find a way to 'manufacture' more points of agreement and this is basically done via induction.  

If $s$ has repeated roots 1st consider the case where only one $\lambda_k$ is not simple and start by removing repeated factors and run the above nice case.  Then multiply each side by one repeated factor [in the denominator] and run the nice case on all entries on the right without repeated factors.  Then multiply by another repeated factor [again in the denominator] and run the nice case on the RHS for all elements without repeated factors and so on getting the result when only one root isn't simple.  Now consider the case where two distinct $\lambda_k$ are not simple but first remove all repeats for one of the roots and use the preceding argument-- then multiply each side by one of the factors and run the preceding argument on the RHS and so on.  For a complete derivation consider page 16 of Marshall's *Complex Analysis* (2.4) in that book reads:  

$\frac{p(z)}{\prod_{j=1}^N (z-z_j)^{n_j}}=r(z)=q(z) + \sum_{j=1}^N\sum_{k=1}^{n_j}\frac{c_{k,j}}{(z-z_j)^k}$ 
and with respect to $x\in \mathscr A$, assuming the $z_j\not \in\sigma(x)$ then  
$\implies \frac{p(x)}{\prod_{j=1}^N (x-z_j)^{n_j}}=r(x)=q(x) + \sum_{j=1}^N\sum_{k=1}^{n_j}\frac{c_{k,j}}{(x-z_j)^k}$  
which we can confirm by multiplying each side by $\prod_{j=1}^N (x-z_j)^{n_j}$ (which is an invertible operation) and each side is a polynomial in $x$.  After algebraic re-arrangement this is necessarily the same polynomial function [it inherits this from the complex scalar case-- multiply each side of that by $\prod_{j=1}^N (x-z_j)^{n_j}$ and do suitable algebraic re-arrangement to see the polynomials are the same as they both sides agree on an infinite number of points hence are the same polynomial function].  

**Conclude:** to integrate any rational function (applied to scalars $z$ or $x \in \mathscr A$) we only need to understand how the integration works with respect to polynomial functions and functions of the form $z\mapsto \frac{1}{(x-z_j)^k}$ since linearity of the integral takes care of the rest.  And for polynomial functions $p(z)=\sum_{k=0}^n \alpha_k \cdot z^k$ we need only understand how the integral works with respect to monomials since there too linearity of the integral takes care of the rest.  





### **(i.) polynomial functions** 
Consider polynomial function $p_m(z)= z^m$ for $m\in \big\{0,1,2,\dotso\big\}$ where $p_0(z)=1$ and when applied to an element of a Banach algebra we mean $p_0(x)=e$.  Let $\Gamma$ be a piecewise $C^1$ closed curve or cycle with minimum modulus $\gt \Vert x\Vert$ that meets the usual winding number criterion [note this implies zero is in the same component as the spectra]; this allows for a Neumann series expansion  argument though of course via Cauchy-Goursat we know we get the same result from  any homologous curve (or cycle).  


Suppose for **contradiction** that $\Big\{x^m,\int_\Gamma  p_m(z)(ze -x)^{-1}dz\Big\}$ are linearly independent.  Then per Hahn-Banach there is some bounded linear functional $\varphi$ such that $\varphi\left(\int_\Gamma p_m(z)(ze - x)^{-1}dz\right) = 1$ and $\varphi\big(x^m\big)=0$ [Corollary 3.1.7 / MacCluer Corollary 3.6 -- applies to the 1 dimensional subspace generated by $x^m$]. This yields    

$$ 1=\varphi\left(\int_\Gamma p_m(z)(ze - x)^{-1} dz\right) = \varphi\left(\int_\Gamma z^m\sum_{k = 0}^\infty z^{-k-1}x^k dz\right)= \int_\Gamma \sum_{k = 0}^\infty z^{m-k-1}\varphi\left( x^k \right)dz=  \sum_{k = 0}^\infty \int_\Gamma z^{m-k-1}\varphi\left(x^k \right)dz =   \int_\Gamma z^{-1}\varphi\big(x^m \big)dz =0$$   
where the interchange of integral and series is the usual result from single variable complex analysis based on uniform convergence. And $1=0$ is of course a contradiction so we deduce $\int_\Gamma  p_m(z)(ze -x)^{-1}dz=\lambda \cdot x^m$. 

Applying Hahn-Banach once more we have bounded linear functional $\psi$ that observes $\psi\big(x^m\big)=1$ which implies  

$$ \lambda= \psi\left(\int_\Gamma p_m(z)(ze - x)^{-1} dz\right) = \psi\left(\int_\Gamma z^m\sum_{k = 0}^\infty z^{-k-1}x^k dz\right)= \int_\Gamma \sum_{k = 0}^\infty z^{m-k-1}\psi\left( x^k \right)dz=  \sum_{k = 0}^\infty \int_\Gamma z^{m-k-1}\psi\left(x^k \right)dz =   \int_\Gamma z^{-1}\psi\big(x^m \big)dz =2\pi i$$   

Since this holds for monomials, then via linearity of the integral it holds for arbitrary polynomials (/polynomial functions) in** $x$    


### (ii.) rational functions   

Let $f(z)= (z-\alpha)^{-m}$ for positive integer $m$ and $\alpha\not \in \sigma(x)$  
and let $\Gamma=(\xi, \gamma)$ be a cycle where $\xi$ is a positively oriented simple circular curve with *very large* radius (which implies $0$ is in the bounded component) and $\gamma$ is a negatively oriented very small circular curve around $\alpha$.  [Again by Cauchy-Goursat, which is implied to hold over operators / elements of a Banach Algebra by Hahn-Banach the integral will be the same over any Homologous cycle.] The integral around $\xi$ evaluates to zero, so we can reduce this to the case of $\Gamma =(\gamma)$.

**Details on why** $ \int_\xi (z-\alpha)^{-m}\big(ze -x\big)^{-1}dz=\mathbf 0$       

Apply arbitrary bounded linear functional $\varphi$ and get  
$$\varphi\left(\int_\xi (z-\alpha)^{-m}\big(ze -x\big)^{-1}dz\right)= \varphi\left(\int_\xi (z-\alpha)^{-m}\sum_{j=0}^\infty z^{-j-1} \cdot x^j dz\right)=\int_\xi (z-\alpha)^{-m}\sum_{j=0}^\infty z^{-j-1} \cdot\varphi\left( x^j\right) dz=\sum_{j=0}^\infty \int_\xi (z-\alpha)^{-m}z^{-j-1} \cdot\varphi\left( x^j\right) dz$$    
and that series is a sum entirely of zeros since for $j\in \mathbb N\cup\big\{0\big\}$   
$$\int_\xi (z-\alpha)^{-m}z^{-j-1} \cdot\varphi\left( x^j\right) dz =  \varphi\left( x^j\right) \cdot\int_\xi \frac{1}{z^{m}}\frac{1}{(1-\frac{\alpha}{z})^{m}}z^{-j-1} dz=\varphi\left( x^j\right)\cdot\int_\xi z^{-m}\left(\sum_{k=0}^\infty \alpha^k\cdot z^{-k}\right)^{m}z^{-j-1}  dz=0$$ 
since e.g. for $m=1$ we have a Laurent series with all negative exponents and zero residue, and for $m\geq 2$, by induction hypothesis we have a Laurent series with all negative exponents and zero residue for $m-1$ and multiply that Laurent series by $\left(\sum_{k=0}^\infty \alpha^k\cdot z^{-k}\right)$ and the Cauchy product results in a Laurent series with all negative exponents and zero residue.  The geometric series exist since $\xi$ is a circular curve around zero with arbitrarily large (enough) radius.  Thus   

$$0=\varphi\left(\int_\xi (z-\alpha)^{-m}\big(ze -x\big)^{-1}dz\right)\implies \mathbf 0=\int_\xi (z-\alpha)^{-m}\big(ze -x\big)^{-1}dz$$    
since $\varphi$ was arbitrary and the dual space separates the points in $\mathscr A$ per Hahn-Banach.

**The evaluation of** $ \int_\gamma (z-\alpha)^{-m}\big(ze -x\big)^{-1}dz=\big(x-\alpha\big)^{-m}$       


As a warm-up, first consider the nice case of $f(z)=z^{-1}$  
$$ f(x)=\frac{1}{2\pi i}\int_\Gamma z^{-1}(ze - x)^{-1} dz=\frac{1}{2\pi i}\int_\gamma z^{-1}(ze - x)^{-1} dz$$  
$$\implies x\cdot f(x)=\frac{1}{2\pi i}\int_\gamma z^{-1}x\cdot (ze - x)^{-1} dz=\frac{-1}{2\pi i}\int_\gamma z^{-1}(-zx^{-1} + e)^{-1} dz=\frac{-1}{2\pi i}\int_\gamma z^{-1} \sum_{j=0}^\infty (-zx^{-1})^j  dz=\frac{-1}{2\pi i}\int_\gamma z^{-1} (-zx^{-1})^0  dz=-1\cdot -1$$ 
The Neumann series expansion holds since $z$, coming from the very small curve, may be made arbitrarily small; note  the series 'disappears' as in the polynomial function argument by use of well chosen $\varphi$ e.g. $\varphi(e)=0$ [for the $j=0$ case] and $\varphi\big(x\cdot f(x)\big)=1$ to generate a contradiction, then $\psi(e)=1$ for the follow-up argument. And the right hand side evaluates to $-1\cdot -1=1$ since $\gamma$ is negatively oriented.    

Now in general for $f(z)= (z-\alpha)^{-m}$ for $m\in \mathbb N$ and $\alpha\not \in \sigma(x)$ and let $w:=x-\alpha e=x-\alpha$ (where the RHS is short-hand)   
$$(x-\alpha )^m\cdot f(x)=w^{m-1}\cdot f(x)=\frac{w^{m-1}}{2\pi i}\int_\gamma (z-\alpha)^{-m}w\cdot \big((z-\alpha)e - w\big)^{-1} dz=  \frac{-1\cdot w^{m-1}}{2\pi i}\int_\gamma (z-\alpha)^{-m} \big(-(z-\alpha)w^{-1} + e\big)^{-1} dz$$ $$= \frac{-1\cdot w^{m-1}}{2\pi i}\int_\gamma (z-\alpha)^{-m}\sum_{j=0}^\infty \big((z-\alpha)w^{-1}\big)^jdz= \frac{-1\cdot w^{m-1}}{2\pi i}\int_\gamma (z-\alpha)^{-1}w^{-(m-1)}dz= \frac{-1}{2\pi i}\int_\gamma (z-\alpha)^{-1}dz =1\implies f(x)=(x-\alpha)^{-m}$$  

where again for any bounded linear functional $\varphi$ we have  
$$\varphi\left(\int_\gamma (z-\alpha)^{-m}\sum_{j=0}^\infty \big((z-\alpha)w^{-1}\big)^jdz\right)=\int_\gamma \sum_{j=0}^\infty (z-\alpha)^{j-m}  \varphi\big( w^{-j}\big)dz=\sum_{j=0}^\infty \int_\gamma  (z-\alpha)^{j-m}  \varphi\big( w^{-j}\big)dz= \int_\gamma  (z-\alpha)^{m-1}  \varphi\big( w^{-(m-1)}\big)dz$$    
where the integral kills all terms in the series except the $j=m-1$ terms [residue] so by the same argument in the polynomial function case   $\left\{ w^{-(m-1)}, \int_\gamma (z-\alpha)^{-m}w\cdot \big((z-\alpha)e - w\big)^{-1} dz\right\}$ are linearly dependent and follow up with $\varphi$ chosen such that $\varphi\big( w^{-(m-1)}\big)=1$ to confirm the answer.

If $f,g\in \mathbf F(x)$ and $\alpha,\beta \in \mathbb C$ then $\alpha f +\beta g\in \mathbf F(x)$ and $(\alpha f +\beta g)(x)= \alpha f(x) +\beta g(x)$  
$$(\alpha f + \beta g)(x)=\frac{1}{2\pi i}\int_\Gamma (\alpha f + \beta g)(\lambda)(\lambda e-x)^{-1}d\lambda=\alpha\frac{1}{2\pi i}\int_\Gamma f (\lambda)(\lambda e-x)^{-1}d\lambda++\beta\frac{1}{2\pi i}\int_\Gamma g (\lambda)(\lambda e-x)^{-1}d\lambda=\alpha f(x) + \beta g(x)$$

**(b.) multiplication**  
If $f,g\in \mathbf F(x)$ then $f\cdot g\in\mathbf F(x)$ and $(f\cdot g)(x)= f(x)\cdot g(x)$     

Using Runge consider rational functions $r_n\to f$ and $s_n \to g$ each uniformly on some compact set $K\supset \sigma(x)$  where cycle $\Sigma$ with $[\Sigma]\in K$ has the winding number criterion implied in the opening statement and $K\subseteq O$, an open set on which $f,g$ are analytic.  Then for all $n\in \mathbb N$   

$r_n(z)\cdot s_n(z)$ is a rational function hence analytic for all $n$ so  
$r_n(x)\cdot s_n(x)$ is well defined as a rational function applied to $x$ (and has poles outside of $K$) hence  

$$r_n(x)s_n(x)=\frac{1}{2\pi i}\int_\Gamma r_n(\lambda)\cdot s_n(\lambda)(\lambda e-x)^{-1}d\lambda$$  
which implies  

$$ f(x)g(x)=\lim_{n\to\infty }r_n(x)s_n(x)=\frac{1}{2\pi i}\int_\Gamma \lim_{n\to\infty } r_n(\lambda)s_n(\lambda)(\lambda e-x)^{-1}d\lambda=\frac{1}{2\pi i}\int_\Gamma f(\lambda)g(\lambda)(\lambda e-x)^{-1}d\lambda=\frac{1}{2\pi i}\int_\Gamma (f g)(\lambda)(\lambda e-x)^{-1}d\lambda=(f g)(x)$$  

Working from right to left, the final equality is the definition of analytic function $fg$ applied to $x$ [recall from [scalar] complex analysis that, subject to Domain agreement, rational functions form a ring], the next equality  is essentially a definition from (scalar) complex analysis, that $(fg)(z)=f(z)g(z)$. The next equality is justified since $r_n(z)\to f(z)$ and $s_n(z)\to g(z)$ uniformly on $[\Gamma]\subseteq K$  and the next equality [i.e. interchange of limit and integral] is justified since that convergence is uniform on $K\supseteq [\Gamma]$ and we are in effect using a Riemann integral (though of course many of the details were left sketchy by the authors while Retherford fills in some of them) and the 1st equality is justified as follows:  
$\big\Vert f(x)g(x)-r_n(x)s_n(x)\big\Vert \leq \big\Vert f(x)g(x)- r_n(x)g(x)\big\Vert +\big\Vert r_n(x)g(x)-r_n(x)s_n(x)\big\Vert\leq \big\Vert f(x)- r_n(x)\big\Vert \big\Vert g(x)\big\Vert +\big\Vert r_n(x)\big\Vert \big\Vert g(x)-s_n(x)\big\Vert\to 0$    
since  $\big\Vert f(x)- r_n(x)\big\Vert\to 0$ and $\big\Vert g(x)- s_n(x)\big\Vert\to 0$-- e.g. as with the middle equality, write out the integral representation of $f(x)-r_n(x)$ and pass a limit, making use of uniform convergence-- and of course $\big\Vert r_n(x)\big\Vert$ is bounded (i.e converges to $\big\Vert f(x)\big\Vert$ by continuity of norms).  Note a nearly identical argument is tacitly used with the uniform convergence justifying the middle equality, i.e. with $\big\Vert h\big\Vert_\infty=\max_{z\in [\Gamma]}\big\vert h(z)\big\vert$ then  

$\big\Vert f\cdot g\cdot -r_n\cdot s_n\big\Vert_\infty \leq \big\Vert f\cdot g- r_n\cdot g\big\Vert_\infty +\big\Vert r_n\cdot g-r_n\cdot s_n\big\Vert_\infty\leq \big\Vert f- r_n\big\Vert_\infty \big\Vert g\big\Vert_\infty +\big\Vert r_n\big\Vert_\infty \big\Vert g-s_n\big\Vert_\infty\to 0$    
(This means the convergence of $r_n\cdot s_n$ to $f\cdot g$ is uniform on $[\Gamma]$ and consider the earlier mentioned 'ML Inequality Implications.)



**optional corollary: composition**  
$\big(f\circ g\big)\big(x\big)=f \big(g(x)\big)$   
for analytic $f,g$ and $\sigma\big(x\big)\in \text{ domain }g$ and $\sigma\big(g(x)\big)\in \text{ domain }f$  
note: here "domain" just refers to domain of a function which is taken to be an open set, not necessarily an open connected set as is common in complex analysis; ref later work for spectral mapping properties, which is independent of this optional result on composition.  With suitable cycles $\Gamma$ and $\Sigma$ chosen (with the winding number criterion, enclosing the relevant spectrum, etc.), let $r_k\to f$ uniformly on a compact set in the domain of $F$ containing $\sigma\big(g(x)\big)$, then    

$$ \big(f\circ g\big)\big(x\big)= \frac{1}{2\pi i}\int_\Gamma f\circ g(\lambda)(\lambda e-x)^{-1}d\lambda=\frac{1}{2\pi i}\int_\Gamma \lim_{k\to\infty}r_k\circ g(\lambda)(\lambda e-x)^{-1}d\lambda= \lim_{k\to\infty}\frac{1}{2\pi i}\int_\Gamma r_k\circ g(\lambda)(\lambda e-x)^{-1}d\lambda= \lim_{k\to \infty}\big(r_k\circ g\big)\big(x\big)$$  
$$ =\lim_{k\to \infty}r_k\big(g(x)\big)= \lim_{k\to\infty}\frac{1}{2\pi i}\int_\Sigma r_k(\lambda)\big(\lambda e-g(x)\big)^{-1}d\lambda=\frac{1}{2\pi i}\int_\Sigma \lim_{k\to\infty} r_k(\lambda)\big(\lambda e-g(x)\big)^{-1}d\lambda=\frac{1}{2\pi i}\int_\Sigma f(\lambda)\big(\lambda e-g(x)\big)^{-1}d\lambda=f \big(g(x)\big)$$

where the interchange of limit and integral is justified on grounds of uniform convergence (optional: use Hahn-Banach to reduce to the single variable complex analysis case).  The key equality is the one between the end of line 1 and the beginning of line 2 which is an algebraic result reflecting the fact that we know for each $k\in \mathbb N$ that  
$\big(r_k\circ g\big)\big(x\big)= r_k\big(g(x)\big)$, which is essentially an algebraic result, and one that we verify below.  


Drawing on the prior discussion of partial fractions, we may write each $r_k$ in the form of   
$q(z) + \sum_{j=1}^N\sum_{m=1}^{n_j}\frac{c_{k,j}}{(z-z_j)^m}$ 
i.e. as a polynomial (which is a linear combination of monomials) and a linear combination of elements of the form $\frac{1}{(z-z_j)^m}$   

We already know from the multiplication property    
$\big(q\circ g\big)\big(x\big)= q\big(g(x)\big)$ since e.g. if $q(z)=p_2(z) =z^2$  then this reads 
$\big(q\circ g\big)\big(x\big)=\big(g \cdot g\big)\big(x\big)= \big(g(x)\big)^2$ and if $q(z)=z^m$  then the result follows by induction, thus in general since $q(z)$ is a linear combination of monomials where the result holds [note we can apply one monomials at a time thanks to linearity of the integral], then it must hold for $q$ as well.  Finally [again by linearity of the integral] it remains to check for the mapping given by $S(z)=\frac{1}{(z-z_j)^m}$, where of course the pole $z_j$ is not in the domain of $f$.  Define $Q(z) = (z-z_j)^m$  


$e =   \big(S\cdot Q\big)\big(g(x)\big) =   S\big(g(x)\big)\cdot Q\big(g(x)\big)= S\big(g(x)\big)\cdot \big(Q\circ g\big)\big(x\big)\quad \text{&} \quad e =    Q\big(g(x)\big)\cdot S\big(g(x)\big)=\big(Q\circ g\big)\big(x\big)\cdot  S\big(g(x)\big)$   
where the final equality holds by the preceding argument for polynomials.  However, for $z$ in our domain $(S\circ g )(z)\cdot (Q\circ g)(z)=\frac{1}{\big(g(z)-z_j\big)^m}\cdot \big(g(z)-z_j\big)^m =1$   
then by applying the multiplication property  
$\big(S\circ g\big)\big(x\big)\cdot \big(Q\circ g\big)\big(x\big)=\Big(\big(S\circ g\big)\cdot \big(Q\circ g\big)\Big)\big(x\big)=e$  
$\implies \big(S\circ g\big)\big(x\big)\cdot \big(Q\circ g\big)\big(x\big) =e = S\big(g(x)\big)\cdot \big(Q\circ g\big)\big(x\big)\implies S\big(g(x)\big)=\big(S\circ g\big)\big(x\big)$ 
since $\big(Q\circ g\big)\big(x\big)$ is invertible.  

Conclude (via linearity of the integral) that $\big(r_k\circ g\big)\big(x\big)= r_k\big(g(x)\big)$    




**(c.) power series implications**   
If the function $f$ is given by a power series [expansion at zero] $f(\lambda )=\sum_{k=0}^\infty \alpha_k \lambda^k$ converging in a neighborhood of $\sigma(x)$ then $f(x) =\sum_{k=0}^\infty \alpha_k x^k$  

Select closed disc $\overline B(0,R)$ with with radius $R=r(x)+2\delta$ [with $r(x)$ being the spectral radius of $x$] for $\delta \gt 0$ small enough such that $f$ converges absolutely (and uniformly) on $\overline B(0,R)$ noting its modulus is dominated by $\sum_{k=0}^\infty \vert \alpha_k\vert \cdot \vert z\vert^k\leq \sum_{k=0}^\infty \vert \alpha_k\vert \cdot R^k$  and let $\Gamma=(\gamma)$  with $\gamma:[0,1]\longrightarrow \mathbb C$ given by $\gamma(t)=\big(r(x)+\delta\big)\cdot\exp\big(2\pi i \cdot t\big)$ which meets the winding number criterion with respect to the spectra of $x$

*Easy case: polynomials*  
We start by considering polynomial functions which are the simplest analytic (in fact entire) functions and mimicking 17.24 (page 270) in Pryce.  The result for constant polynomials trivially holds, so consider polynomial functions with degree $n\geq 1$.   

$p(z) = \prod_{k=1}^n \alpha \cdot(z-\lambda_k)$ and consider for $x$ in our Banach algebra    

$0\not\in \sigma\big(p(x)\big)\iff p(x)\text{ invertible}\iff \text{ each }(x-\lambda_k e)\text{ invertible}\iff \text{ each }\lambda_k\not\in\sigma(x)\iff 0\not\in p\big(\sigma(x)\big)$   
for the 2nd implication consider that if some $(x-\lambda_j e)$ is not invertible for some $j$ then $P\big(x\big)= \big(\alpha\prod_{k\neq j}(x-\lambda_k e)\big)\big(x-\lambda_j e\big)= \big(x-\lambda_j e\big)\big(\alpha\prod_{k\neq j}(x-\lambda_k e)\big)$  
where the middle representation is not injective when $
\big(x-\lambda_j e\big)$ is not injective [under the 'product' operation] and the right hand side is not surjective when $\big(x-\lambda_j e\big)$ is not surjective since $\text{image }\Big(\big(x-\lambda_j e\big)\big(\alpha\prod_{k\neq j}(x-\lambda_k e)\big)\Big)\subseteq \text{image }\big(x-\lambda_j e\big)$   

Now get the general result by translation with $P(z)=p(z)-\lambda$ for some $\lambda \in \mathbb C$ and this becomes   
$\lambda \not\in \sigma\big(p(x)\big)\iff 0\not\in \sigma\big(p(x)-\lambda\big)\iff 0\not\in \sigma\big(P(x)\big)\iff 0\not\in P\big(\sigma(x)\big)\iff 0\not\in \big(p\big(\sigma(x)\big)-\lambda\big)\iff \lambda \not\in p\big(\sigma(x)\big)$  
$\implies \lambda \in \sigma\big(p(x)\big)\iff \lambda \in p\big(\sigma(x)\big)$

**A much nicer approach for spectral mapping, using maximal commutative sub-algebras**     
$\zeta \in \sigma\big(f(x)\big)\iff \zeta \in f\big(\sigma(x)\big)$  

Now let $S=\big\{x\big\}$ and $\mathscr B =\Upsilon\big(\Upsilon(S)\big)$ [ref ex 10.9.23 for details] which is a *commutative* [unital] Banach (hence closed) sub-algebra of $\mathscr A$ where necessarily $x\in S\subseteq \mathscr B$ and $\sigma_{\mathscr B}(x)=\sigma_{\mathscr A}(x)$ [part b then c of ex 10.9.23]  

Thus we can resort to using multiplicative linear functionals $\varphi \in \Delta =\mathscr M_{\mathscr B}$ [the latter being better notation albeit from MacCluer]  

Note $f(x) = \lim_{k\to \infty}r_k(x)$ where each $r_k(x)\in \mathscr B$ [ref commentary in ex 10.9.23(c)] hence the limit $f(x)\in \mathscr B$ as well since $\mathscr B$ is closed.  [We could instead make the argument on the integral itself being a limit of a Riemann sum of elements in $\mathscr B$ if we wanted.]  So we may restrict to the sub algebra $\mathscr B$ and  

**via rational functions**  
$\sigma\big(f(x)\big)\subseteq f\big(\sigma(x)\big)$  
Consider  $\zeta \in \sigma_{\mathscr B}\big(f(x)\big)=\sigma_{\mathscr A}\big(f(x)\big)$.  Then for some multiplicative linear functional $\varphi \in \mathscr M_{\mathscr B}$ we have  
$\zeta=\varphi\big( f(x)\big)=\varphi\big(\lim_{k\to\infty} r_k(x)\big)=\lim_{k\to \infty} \varphi\big( r_k(x)\big)=\lim_{k\to \infty} r_k\big( \varphi(x)\big)=f\big(\varphi(x)\big)$  
$\implies \zeta =f\big(\varphi(x)\big)\in f\Big(\sigma_{\mathscr A}\big(x\big)\Big)$ 

$\sigma\big(f(x)\big)\supseteq f\big(\sigma(x)\big)$  
Consider $\zeta \in f\Big(\sigma_{\mathscr B}\big(x\big)\Big)=f\Big(\sigma_{\mathscr A}\big(x\big)\Big)$, then $\zeta = f\big(\varphi(x)\big)$ for some $\varphi \in \mathscr M_{\mathscr B}$ so running the argument backwards  
$f\big(\varphi(x)\big)  =\lim_{k\to \infty} r_k\big( \varphi(x)\big) =  \lim_{k\to \infty} \varphi\big( r_k(x)\big)= \varphi\big(\lim_{k\to\infty} r_k(x)\big)= \varphi\big( f(x)\big)=\zeta$  

**note:**  
This argument hinges on the fact that we know how a multiplicative linear functional behaves on rational functions of $x$ then takes limits.  It is conceptually the same as when working with a $^*$-homomorphism and a commutative unital $C^*$-algebra.  In that case we work with polynomials of an operator [viewed as self-adjoint / generalized polynomials $p(z,\overline z)$ or multivariate polynomials of elements in the algebra] and again have $p(x)\to f(x)$ for some continuous $f$ and get the spectral mapping property by using multiplicative linear functionals.   

*The same argument, but this time factored through the Cauchy Integral Formula*  
$\implies$   
if $\zeta \in \sigma_{\mathscr A}\big(f(x)\big)=\sigma_{\mathscr B}\big(f(x)\big)$ then there is some multiplicative linear functional $\varphi \in \Delta =\mathscr M_{\mathscr B}$  [this is Theorem 5.28 in MacCluer], so by the (scalar) Cauchy Integral Formula  




$$\zeta = \varphi\big(f(x)\big)=\varphi\left(\frac{1}{2\pi i}\int_\Gamma f(\lambda)(\lambda e-x)^{-1}d\lambda\right)=\frac{1}{2\pi i}\int_\Gamma f(\lambda)\varphi\big((\lambda e-x)^{-1}\big)d\lambda=\frac{1}{2\pi i}\int_\Gamma \frac{f(\lambda)}{\lambda -\varphi(x)} d\lambda =f\big(\varphi(x)\big)\in f\big(\sigma_{\mathscr A}(x)\big)$$    
since $\varphi(x)\in \sigma_{\mathscr B}\big(x\big)=\sigma_{\mathscr A}\big(x\big)$  

$\impliedby$  
Running the argument in reverse, let $\zeta =\varphi\big(x\big)\in \sigma_{\mathscr B}\big(x\big)=\sigma_{\mathscr A}\big(x\big)$ then  

$$f\big(\zeta\big)=f\big(\varphi(x)\big) = \frac{1}{2\pi i}\int_\Gamma \frac{f(\lambda)}{\lambda -\varphi(x)} d\lambda=  \varphi\big(f(x)\big)=\frac{1}{2\pi i}\int_\Gamma f(\lambda)\varphi\big((\lambda e-x)^{-1}\big)d\lambda=\varphi\left(\frac{1}{2\pi i}\int_\Gamma f(\lambda)(\lambda e-x)^{-1}d\lambda\right)= \varphi\big(f(x)\big)$$  

thus $f\big(\zeta\big)=\varphi\big(f(x)\big)\in \sigma_{\mathscr B}\big(f(x)\big)=\sigma_{\mathscr A}\big(f(x)\big)$

Let $\mathscr A$ be a commutative [necessarily unital per page 167] semi-simple Banach algebra with involution $x\mapsto x^*$.  Prove that the involution is continuous in $\mathscr A$, i.e. if $x_n\to x$ then $x_n^*\to x^*$  
[Hint: for any multiplicative functional $h\in \Delta$ consider the functional $\varphi(x) = \overline{h(x^*)}$, and then use the Closed Graph Theorem]  


- - - -   
note: $\Delta$ is defined on page 180, seemingly as the set of multiplicative linear functionals [excluding the zero functional]-- this would be $\mathscr M_{\mathscr A}$ in MacCluer's notation, which I prefer somewhat.   


another note: per theorem 10.5.3, being semi-simple means the only element with $\sigma(x)=\big\{0\big\}$ is the zero element (quasi-nilpotent in terms of MacCluer ex 5.15), which means the multiplicative linear functionals can separate the points of $\mathscr A$ since $x-y$ is zero under all multiplicative linear functionals *iff* all spectra is zero, *iff* it is the zero element by semi-simplicity.

That $\varphi(x) = \overline{h(x^*)}$ is a linear functional is fairly immediate.  It is multiplicative since  
$\varphi (xy) = \overline{h((xy)^*)}=\overline{h(y^*x^*)}=\overline{h(y^*)}\cdot \overline{h(x^*)}=\overline{h(x^*)}\cdot \overline{h(y^*)}=\varphi (x)\cdot \varphi (y) $  
and of course $\varphi(e) =1$ so it has norm $1$ by the usual arguments of $\varphi \left(e -\varphi(x)^{-1}\cdot x\right)$ for any $x$ such that $\varphi(x)\neq 0$.  Thus $\Vert \varphi\Vert =1$.  

Consider the mapping $T:\mathscr A\longrightarrow \mathscr A$ given by $T(x)=x^*$. It is an *anti-linear* operation, however while the authors assert that that we can still use the Closed Graph Theorem for anti-linear maps between Banach spaces, they do not really explain why and it does not seem to obviously follow from the the standard proof of Closed Graph Theorem via Inverse Mapping Theorem [Remark 9.3.3 on page 134 or Theorem 3.21 in MacCluer].  It may follow from the author's proof of Theorem 9.3.2 via perfect convexity, though I found that argument hard to follow, so **some major technical issues remain here**.  Crucially, anti-linear map respects the underlying abelian group operation, hence respects convergent sequences.  

So, in line with the Closed Graph Theorem (and in particular MacCluer ex 3.21), it suffices to consider the case where $x_n\to 0$ and $x_n^*=Tx_n\to b$, and we need to show $b=\mathbf 0$.  Each multiplicative linear functional is continuous [with norm $1$] and for arbitrary $h\in \Delta$, working from right to left      
$0=\varphi\big(\mathbf 0\big)=\varphi\big(\lim_{n\to\infty} x_n\big)= \lim_{n\to\infty}\overline{h\big( x_n^*\big)}=\overline{h\big(\lim_{n\to\infty} x_n^*\big)}=\overline{h\big(b\big)}$  
$\implies \sigma\big(b\big)=\big\{0\big\}\implies b=\mathbf 0$  
Conclude $T$ is continuous by Closed Graph Theorem.  


*remark:*  
In terms of motivation: the suggested $\varphi$ to use is reminiscent of $\overline {f(\overline z)}$ for analytic $f(z)$ in complex analysis.  Also heuristically, the approach here is, to show $T$ is bounded, since it is anti-linear, we 'want' to make it linear by considering $\overline{Tx}$ but 'complex conjugation' of an element in $\mathscr A$ is is not defined... however when we switch to linear functionals, the co-domain is $\mathbb C$ so $\overline {h(Tx)}$ is sensible. If we had tried to use the general dual space, i.e.  $h\in\mathscr A^*$, then we wouldn't know that $\varphi$ is bounded; however since the space is semi-simple, multiplicative linear functionals separate the points and restricting to $h\in \Delta$ then tells us that $\varphi$ is multiplicative hence bounded.  

**A fix for the above**   
One fix for the above is to recognize that the set of all multiplicative linear functionals on $\mathscr A$ is **total** [ref problem 9D in Pryce] by semi-simplicity, i.e. that collection of functionals separates the points of $\mathscr A$.  This implies that $x\in \mathscr A$ is zero *iff* $\text{Re }(x)=0=\text{Im }h(x)$ for all $h \in \mathscr M_{\mathscr A}$ i.e. all linear functionals in that collection.  

So let $V$ be the Banach space of $\mathscr A$ with the restriction of only allowing $\mathbb R$ scalars and as above $T:V\rightarrow V$ is given by $Tx=x^*$ which is a linear map on $V$.  Using Closed Graph Theorem / MacCluer ex 3.21 we consider $x_n\to 0$ and $Tx_n \to b$.  Then consider elements from our total sets, i.e. for $h \in \mathscr M_{\mathscr A}$ we have  

$\text{Re  }h\big(b\big)=\lim_{n\to\infty}\text{Re  }h\big(Tx_n\big)=\lim_{n\to\infty}\text{Re  }h\big(x_n^*\big)=\lim_{n\to\infty}\text{Re  }\overline{h\big(x_n^*\big)}=\lim_{n\to\infty}\text{Re  }\varphi\big(x_n\big)=\text{Re  }\varphi\big(\lim_{n\to\infty} x_n\big)=\text{Re  }\varphi\big(\mathbf 0\big)=0$   
$\text{Im  }h\big(b\big)=\lim_{n\to\infty}\text{Im  }h\big(Tx_n\big)=\lim_{n\to\infty}\text{Im  }h\big( x_n^*\big)=\lim_{n\to\infty}-\text{Im  }\overline{h\big( x_n^*\big)}=\lim_{n\to\infty}-\text{Im  }\varphi\big(x_n\big)= -\text{Im  }\varphi\big(\lim_{n\to\infty}x_n\big)= -\text{Im  }\varphi\big(\mathbf 0\big)=0$   
Noting that $\text{Re  }h$ and $\text{Im  }h$ both specify linear functionals on $V$ with norm (at most) $=1$ since $h\in \mathscr M_{\mathscr A}$ so $\Vert h\Vert =1$.  

**A different argument**   
If $a_n\to a_0\in \mathscr A$ where $\big\{a_n\big\}$ are self-adjoint then $a_0$ is self-adjoint.  Consider (multiplicative linear functional) $h \in \mathscr M_{\mathscr A}$ and (multiplicative linear functional) $\varphi(x) := \overline{h(x^*)}$ as in the preceding argument.  Now  $\mathbb R \ni h\big(a_n)\to h(a_0)$ so $h(a_0)\in \mathbb R$ and similarly $\mathbb R \ni \varphi\big(a_n)\to \varphi(a_0)$ so $\varphi(a_0)\in \mathbb R$ and $\overline\varphi(a_0)\in \mathbb R$  thus  
$h\left(\frac{a_0- a_0^*}{2}\right) = h\left(\frac{a_0}{2}\right)-h\left(\frac{a_0^*}{2}\right)= \frac{1}{2}\Big(h\left(a_0\right)-\overline \varphi\left(a_0\right)\Big)\in \mathbb R\implies \sigma\left(\frac{a_0- a_0^*}{2}\right)\in \mathbb R\cap i\mathbb R\implies\frac{a_0- a_0^*}{2}=\mathbf 0$   

so the spectrum of this element is real but skew elements have purely imaginary spectrum hence the skew element is the zero element by semi-simplicity.   And for skew $b_n\to b_0\in \mathscr A$ we have a nearly identical argument that   

$h\left(\frac{b_0- b_0^*}{2}\right) = h\left(\frac{b_0}{2}\right)-h\left(\frac{b_0^*}{2}\right)= \frac{1}{2}\Big(h\left(b_0\right)-\overline \varphi\left(b_0\right)\Big)\in i\mathbb R\implies \sigma\left(\frac{b_0+ b_0^*}{2}\right)\in i\mathbb R\cap \mathbb R\implies\frac{b_0 + b_0^*}{2}=\mathbf 0$   

This means the sets (or indeed abelian groups) of self-adjoint and skew elements are each closed subsets of $\mathscr A$.  (They are not closed *subspaces* of $\mathscr A$ since e.g. being self-adjoint is not preserved when scaled by a non-real number, though we remedy this momentarily.)  From now on we ignore all notions of spectrum and even the dual-space.  Instead we take $\mathscr A$ and use its elements, abelian group operation ("$+$"), and norm  and **restrict** the scalar field to be $\mathbb R$-- and call this resulting Banach space $V$.  This real Banach space $V$ can can be written as a direct sum of self-adjoint and skew elements, each of which is a closed [equivalently: complete] subspace. Notice that our overall vector space being complete and those two subspaces being complete only depends on the abelian group operation of the vector space as well as the norm -- it does not depend on the scaling operation so switching from $\mathbb C$ to $\mathbb R$ does not alter completeness.  Define $P:V\rightarrow V$ given by  

$P:= \frac{I+T}{2}$   
which observes $P^2 =P$ and in particular $\text{image }P=\text{subspace of self-adjoint elements}$ and $\ker P=\text{subspace of skew elements}$, each of which are closed per the above and this implies $P$ is bounded [Pryce ex 10T, a consequence of Closed Graph Theorem].  

Then $T=2P-I$ is bounded since it is a linear combination of bounded maps.  I.e. for any $x\in \mathscr A$ which is the same element $x\in V$, we have  
$\big\Vert Tx\big\Vert \leq M\lt \infty$.  But $x\in \mathscr A$ and $Tx\in \mathscr A$ are the same elements as $x\in V$ and $Tx \in V$, with the same norm hence $T$ is a bounded anti-linear map over our original complex Banach algebra $\mathscr A$ and if $x_n\to x_0$ then  
$\big\Vert x_n^* -x_0^*\big\Vert=\big\Vert Tx_n -Tx_0\big\Vert=\big\Vert T\big(x_n -x_0\big)\big\Vert \leq \big\Vert T\big\Vert \cdot \big\Vert x_n-x_0\big\Vert =M \cdot \big\Vert x_n-x_0\big\Vert \to 0$ as required.
