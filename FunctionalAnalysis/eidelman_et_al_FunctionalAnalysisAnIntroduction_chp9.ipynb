$x_n:= \sum_{k=1}^n w_k \cdot v_k +\big(\sum_{k=n+1}^\infty w_k\big)\cdot v_{n+1}\in S$   
then for $n\gt m\geq N$, by triangle inequality we have  

$\big\Vert x_n-x_m\big\Vert\leq \big\Vert \sum_{k=m+1}^\infty w_k \cdot v_{m+1}\big\Vert+\big\Vert \sum_{k=m+1}^nw_k \cdot v_{m+1}\big\Vert+ \big\Vert \sum_{k=n+1}^\infty w_k \cdot v_{m+1}\big\Vert$  
$\leq M\cdot \big( \sum_{k=m+1}^\infty w_k +\sum_{k=m+1}^n w_k + \sum_{k=n+1}^\infty w_k \big)\leq 3M\cdot  \sum_{k=m+1}^\infty w_k$ 
where $\sum_{k=m+1}^\infty w_k\to 0^+$ monotone in $N$ hence this is a Cauchy sequence so the limit $L$ exists since the space is Banach   

Similarly  
$\big\Vert x_n -\sum_{k=1}^\infty w_k\cdot v_k \big\Vert \leq \big\Vert \sum_{k=n+1}^\infty w_k \cdot v_{k}\big\Vert+ \big\Vert \sum_{k=n+1}^\infty w_k \cdot v_{m+1}\big\Vert\leq 2M\cdot \sum_{k=n+1}^\infty w_k \cdot v_{m+1}$  
$\implies L =\sum_{k=1}^\infty w_k\cdot v_k$  

and $x_n\in S$ for all $n$ hence its limit is as well since $S$ is closed

Prove the equalities

Translation $T(v)=v-x$ is a homeomorphism so consider sequence $v_k \in K$ converging to $v\in \overline K$ [metric spaces are first countable] then $T(v_k) \in K_x$ for all $k$ and then $T(v_k)\to v-x\in \overline {(K_x})$ by closure hence $\overline{(K_x)}\supseteq \overline K -x$.

$\stackrel{o}{(K_x)} = \big(\overline{(K_x^C)}\big)^C= \big(\overline{(K^C)-x}\big)^C= \big(\overline{(K^C)}-x\big)^C = \overline{(K^C)}^C-x = \stackrel{o}{K}-x$

**16.** Let $\mu$ be a perfectly convex function on a Banach space $X$ such that $\mu(\lambda \cdot x)=\lambda \cdot\mu(x)$ for any $\lambda \geq 0$.  Prove that $\mu$ is a continuous function.  

A perfectly convex function is a real non-negative valued function that observes Jensen's inequality for discrete/countable non-negative weights/probabilities [Def 9.4.1].  Theorem 9.4.2 tells us that for $x \in X$ we have  
$\big \vert \mu(x)\big\vert =\mu(x)\leq C \cdot \big \Vert x\big\Vert$ for some (finite) $C\geq 0$ so $\mu$ is bounded though perhaps not known to be linear so a little more work is needed show continuity.  Now consider some sequence $v_k\to 0$.  Then the above boundedness condition yields  
$\big \vert \mu(v_k)\big\vert \leq C \cdot \big \Vert v_k\big\Vert\to 0$   
$\implies \mu$ is continuous at zero.  

More generally consider sequence $x_k\to x$.  Define $v_k:= x_k-x$ and by manipulating what amounts to triangle inequality and reverse triangle inequality  
$\mu(x_k)=\mu(x + v_k)\leq \mu(x)+\mu(v_k)\iff \frac{1}{2}\cdot\mu (x_k)=\mu \left(\frac{1}{2}\cdot x_k\right)=\mu\left(\frac{1}{2}\cdot x + \frac{1}{2}\cdot v_k\right)\leq \frac{1}{2}\cdot\mu\left(x\right)+\frac{1}{2}\cdot\mu\left(v_k\right)$  
$\mu(x)=\mu(x_k + (-v_k))\leq \mu(x_k) + \mu(-v_k)\iff \frac{1}{2}\cdot\mu (x)=\mu \left(\frac{1}{2}\cdot x\right)=\mu\left(\frac{1}{2}\cdot x_k + \frac{1}{2}\cdot (-v_k)\right)\leq \frac{1}{2}\cdot\mu\left(x_k\right)+\frac{1}{2}\cdot\mu\left(-v_k\right)$    
$\implies \mu(x)-\mu(-v_k)\leq \mu(x_k)\leq \mu(x)+\mu(v_k)$  
$\implies \mu(x) -0 =\lim_{k\to \infty}\big(\mu(x)-\mu(-v_k)\big)\leq \lim_{k\to \infty}\mu(x_k)\leq \lim_{k\to \infty}\big(\mu(x)+\mu(v_k)\big)=\mu(x)+0$  
Thus by squeeze theorem $\mu(x_k)\to \mu(x)$ and $\mu$ is continuous at $x$.

**(ii.)** $\text{ weakly holomorphic}\implies \text{strongly holomorphic }$  


Hint: using Cauchy Integral Formula prove that for any scalar analytic function $g(z)$  
$\left\vert \frac{1}{\alpha-\beta}\cdot \left(\frac{1}{\alpha}\cdot\left( g(z_0+\alpha)-g(z_0)\right)\right)-\left(\frac{1}{\beta}\cdot \left(g(z_0+\beta)-g(z_0)\right)\right)  \right \vert\leq M(g,r)$  
where $z_0\in D$, $\vert \alpha\vert,\vert \beta\vert\lt r \big[$technically this should be distinct $\alpha, \beta\in B\big(0,r\big)$ where one of the values being zero is allowable as it induces a removable singularity we can define over$\big]$and $\big\{\vert z-z_0\vert \lt 2r\big\}\subset D$ holds and then use the corollary from Banach-Steinhaus theorem. 
- - - - -  
Note: the problem statement's hint and even the official solution is written in a confusing manner that seems to indicate a bound $M(g, r)$ that does not depend on $z_0$ which is nonsensical.  Instead the argument the authors wanted was to fix some point $z_0$ and show that being analytic in that neighborhood implies differentiability at $x\big(z_0\big)$. I found using the Cauchy Integral Formula to be a bit unpleasant here and I worked through a power series approach instead, which is more in line with Beardon's *Complex Analysis*.  Since $z_0$ is some fixed value, via translation we can assume WLOG $z_0=0$ [which simplifies notation].    
- - - - -  
consider the power series expansion of $g$ at $0$   
$g(z)=\sum_{k=0}^\infty c_k\cdot z^k$  

$\begin{align}\\  
&\left\vert\frac{1}{\alpha-\beta}\right\vert\cdot\left \vert \frac{g(\alpha)-g(0)}{\alpha} -\frac{g(\beta)-g(0)}{\beta}\right\vert\\  
&\left\vert\frac{1}{\alpha-\beta}\right\vert\cdot\left \vert \sum_{k=0}^\infty c_{k+1}\cdot \big(\alpha^k -\beta^k\big)\right\vert\\  
&\left\vert\frac{1}{\alpha-\beta}\right\vert\cdot\left \vert \sum_{k=1}^\infty c_{k+1}\cdot \big(\alpha^k -\beta^k\big)\right\vert\quad\text{ (since }\alpha^0-\beta^0=0\text{)}\\  
&=\left \vert \sum_{k=1}^\infty c_{k+1}\cdot \sum_{j=0}^{k-1}\alpha^j \cdot \beta^{k-1-j}\right\vert\quad\text{(1.2.2 on page 8 of Beardon)}\\  
&\leq \sum_{k=1}^\infty \vert c_{k+1}\vert \cdot \sum_{j=0}^{k-1}\vert \alpha\vert^j \cdot \vert \beta\vert^{k-1-j}\\  
&\leq \sum_{k=1}^\infty \vert c_{k+1}\vert \cdot \sum_{j=0}^{k-1}r^{k-1}\\  
&= \sum_{k=1}^\infty k\cdot\vert c_{k+1}\vert \cdot r^{k-1}\\
&\leq \sum_{k=1}^\infty (k+1)\cdot\vert c_{k+1}\vert \cdot r^{k-1}\\
\end{align}$    

where the upper bound converges in the radius of convergence (stated to be at least $2r$) since if we scale by $r$ we recover  

$\sum_{k=1}^\infty (k+1)\cdot\vert c_{k+1}\vert \cdot r^{k}\leq \sum_{k=0}^\infty (k+1)\cdot\vert c_{k+1}\vert \cdot r^{k}$ while $g'(z) =\sum_{k=0}^\infty (k+1)\cdot c_{k+1} \cdot z^{k}$  
which converges absolutely at $z:=r$ (i.e. the upper bound on the left)   

Thus for fixed $z_0$ and any complex analytic $g$ we have a bound on $\left\vert\frac{1}{\alpha-\beta}\right\vert\cdot\left \vert \frac{g(\alpha)-g(0)}{\alpha} -\frac{g(\beta)-g(0)}{\beta}\right\vert$ that depends only on $g$ and $r$-- not on the selection of $\alpha, \beta$ that are within $r$ distance of $z_0$.
